```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox, scrolledtext
from PIL import Image, ImageTk
import os
import pandas as pd
import cv2
import numpy as np
import matplotlib.pyplot as plt
import csv
import warnings
from sklearn.cluster import KMeans
from skimage import measure
from scipy.spatial import cKDTree
from scipy import stats
import pyvista as pv

# At the top, add:

class FoamAnalyzerGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("HDPE-HGM Syntactic Foam Analyzer")
        self.root.geometry("1600x1000")
        style = ttk.Style(root)
        style.configure('TFrame', background='#f0f4f8')
        style.configure('TLabel', font=('Times New Roman', 12), background='#f0f4f8')
        style.configure('TButton', font=('Times New Roman', 12))
        style.configure('TEntry', font=('Times New Roman', 12))

        self.annotated_images = []
        self.thresholds_dict = {}

        self.setup_ui()
    
    def setup_ui(self):
        # Top controls frame
        top = ttk.Frame(self.root)
        top.pack(fill='x', pady=5, padx=10)

        # Image folder selection
        ttk.Label(top, text="Image Folder:").grid(row=0, column=0, sticky='w', padx=5)
        self.img_entry = ttk.Entry(top, width=70)
        self.img_entry.grid(row=0, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_image).grid(row=0, column=2, padx=5)

        # CSV file selection
        ttk.Label(top, text="Threshold CSV:").grid(row=1, column=0, sticky='w', padx=5)
        self.csv_entry = ttk.Entry(top, width=70)
        self.csv_entry.grid(row=1, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_csv).grid(row=1, column=2, padx=5)

        # Process button
        ttk.Button(top, text="Process", command=self.process_images).grid(row=2, column=1, pady=10)

        # Progress bar
        self.progress = ttk.Progressbar(self.root, mode='indeterminate')
        self.progress.pack(fill='x', pady=5, padx=10)

        # Middle frame for results and controls
        middle_frame = ttk.Frame(self.root)
        middle_frame.pack(fill='both', expand=True, pady=10, padx=10)

        # Left panel for image viewing
        left_panel = ttk.Frame(middle_frame)
        left_panel.pack(side='left', fill='both', expand=True, padx=5)

        # Threshold labels
        threshold_frame = ttk.Frame(left_panel)
        threshold_frame.pack(fill='x', pady=5)
        
        self.t1label = ttk.Label(threshold_frame, text="Global T1: Not calculated")
        self.t1label.pack(side='left', padx=10)
        
        self.t2label = ttk.Label(threshold_frame, text="Global T2: Not calculated")
        self.t2label.pack(side='left', padx=10)

        # Image display
        image_frame = ttk.Frame(left_panel)
        image_frame.pack(fill='both', expand=True, pady=5)
        
        self.ct_img_label = ttk.Label(image_frame, text="No images loaded", 
                                     relief='sunken', borderwidth=2)
        self.ct_img_label.pack(fill='both', expand=True)

        # Image slider
        slider_frame = ttk.Frame(left_panel)
        slider_frame.pack(fill='x', pady=5)
        
        ttk.Label(slider_frame, text="Image:").pack(side='left')
        self.slider = ttk.Scale(slider_frame, from_=0, to=0, orient='horizontal',
                               command=self.update_image)
        self.slider.pack(side='left', fill='x', expand=True, padx=5)

        # Right panel for 3D visualization and controls
        right_panel = ttk.Frame(middle_frame)
        right_panel.pack(side='right', fill='both', expand=True, padx=5)

        # 3D visualization buttons
        button_frame = ttk.Frame(right_panel)
        button_frame.pack(fill='x', pady=5)
        
        ttk.Button(button_frame, text="Show Voids 3D", 
                  command=self.show_voids3d).pack(side='left', padx=5)
        ttk.Button(button_frame, text="Show Particles 3D", 
                  command=self.show_particles3d).pack(side='left', padx=5)

        # 3D display area
        self.static_3d_label = ttk.Label(right_panel, text="3D Visualization Area", 
                                        relief='sunken', borderwidth=2)
        self.static_3d_label.pack(fill='both', expand=True, pady=5)

        # Volume fraction display
        self.vflabel = ttk.Label(right_panel, 
                                text="VF% - Process images to see results",
                                font=('Times New Roman', 10))
        self.vflabel.pack(fill='x', pady=5)
        
    def browse_image(self):
        """Browse for image folder."""
        path = filedialog.askdirectory(title="Select Image Folder")
        if path:
            self.img_entry.delete(0, 'end')
            self.img_entry.insert(0, path)

    def browse_csv(self):
        """Browse for CSV file."""
        path = filedialog.askopenfilename(
            title="Select Threshold CSV File",
            filetypes=[("CSV files", "*.csv")]
        )
        if path:
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, path)

    def process_images(self):
        """Process images and create thresholds."""
        folder = self.img_entry.get()
        csv_file = self.csv_entry.get()
        
        if not folder:
            messagebox.showerror("Error", "Please select an image folder")
            return

        # If CSV missing, prompt for save location
        if not csv_file:
            csv_file = filedialog.asksaveasfilename(
                defaultextension=".csv",
                filetypes=[("CSV files", "*.csv")],
                title="Save thresholds CSV"
            )
            if not csv_file:
                messagebox.showerror("Error", "Threshold CSV file path required")
                return
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, csv_file)

        # Start progress bar
        self.progress.start()
        self.root.update()

        try:
            # Load and process images
            images = load_images_from_folder(folder)
            if not images:
                raise ValueError("No valid images found in the selected folder")
            
            # Process images and save results
            process_images_and_save_results(images, "histograms", csv_file)
            
            # Load thresholds
            threshold_result = load_thresholds_from_csv(csv_file)
            if threshold_result[0] is None:
                raise ValueError("Failed to load thresholds from CSV")
            
            self.thresholds_dict, g1, g2 = threshold_result
            self.t1label.config(text=f"Global T1: {g1:.2f}")
            self.t2label.config(text=f"Global T2: {g2:.2f}")

            # Load annotated images for display
            self.annotated_images = []
            for filename, img in images:
                # Convert to display format
                if len(img.shape) == 3:
                    display_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                else:
                    display_img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
                
                pil_img = Image.fromarray(display_img).resize((400, 400))
                self.annotated_images.append(pil_img)
            
            # Update slider and display first image
            if self.annotated_images:
                self.slider.config(to=len(self.annotated_images)-1)
                self.update_image(0)
            
            messagebox.showinfo("Success", "Image processing completed successfully!")
            
        except Exception as e:
            messagebox.showerror("Error", f"Processing failed: {str(e)}")
        finally:
            self.progress.stop()

    def update_image(self, val):
        """Update displayed image based on slider value."""
        try:
            idx = int(float(val))
            if 0 <= idx < len(self.annotated_images):
                img = self.annotated_images[idx]
                imgtk = ImageTk.PhotoImage(img)
                self.ct_img_label.config(image=imgtk, text="")
                self.ct_img_label.image = imgtk  # Keep a reference
        except (ValueError, IndexError):
            pass  # Ignore invalid values

    def show_voids3d(self):
        """Generate and display 3D void visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="")
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")

    def show_particles3d(self):
        """Generate and display 3D particle visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="Particles 3D", compound='top')
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")


    def get_root():
        global _root
        if _root is None:
            _root = tk.Tk()
            _root.withdraw()
        return _root
# Constants
MICRONS_PER_PIXEL = 4.89
SLICE_THICKNESS_MICRONS = 4.89
MAX_VOID_SIZE_MICRONS = 50.0
MAX_VOID_SIZE_PIXELS = MAX_VOID_SIZE_MICRONS / MICRONS_PER_PIXEL
MAX_VOID_AREA_PIXELS = np.pi * (MAX_VOID_SIZE_PIXELS / 2) ** 2
def select_folder(prompt):
    root = get_root()
    folder_path = filedialog.askdirectory(title=prompt)
    if folder_path:
        print(f"{prompt}: {folder_path}")
        return folder_path
    print("No folder selected.")
    return None

def select_csv_file(prompt):
    root = get_root()
    file_path = filedialog.asksaveasfilename(
        title=prompt, 
        defaultextension=".csv",
        filetypes=[("CSV files", "*.csv")]
    )
    if file_path:
        print(f"{prompt}: {file_path}")
        return file_path
    print("No file location selected.")
    return None
def load_images_from_folder(folder_path):
    """Load images from the specified folder."""
    images = []
    for filename in sorted(os.listdir(folder_path)):
        if filename.lower().endswith(('.tif', '.tiff', '.png', '.jpg', '.jpeg')):
            img_path = os.path.join(folder_path, filename)
            img = cv2.imread(img_path)
            if img is not None:
                images.append((filename, img))  # Append as (filename, image)
    return images

def calculate_histogram(gray_image):
    """Calculate and return the histogram of a grayscale image."""
    histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
    return histogram

def calculate_epsilon(histogram, sample_size=10000):
    """Compute epsilon for neighborhood initialization (approximate mean pairwise distance)."""
    pixel_values = np.arange(256)
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot calculate epsilon.")
    
    probabilities = histogram / histogram_sum  # Normalize histogram to get probabilities
    sample_size = int(min(sample_size, histogram_sum))

    sampled_data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    mean_distance = np.mean(np.abs(np.diff(np.sort(sampled_data))))
    return mean_distance

def initialize_kmeans_centers(histogram, k, epsilon=None):
    pixel_values = np.arange(len(histogram))
    histogram_sum = histogram.sum()

    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot initialize K-means centers.")

    probabilities = histogram / histogram_sum  # Normalize histogram
    sample_size = int(min(10000, histogram_sum))
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    if epsilon is None:
        epsilon = calculate_epsilon(histogram)

    centers = []
    for _ in range(k):
        if len(centers) == 0:
            first_center = np.argmax(histogram)
            centers.append(first_center)
        else:
            distances = np.min([np.abs(data - center) for center in centers], axis=0)
            if np.max(distances) <= epsilon:
                new_center = data[np.argmax(distances)]
            else:
                new_center = data[np.argmax(distances)]
            if new_center not in centers:
                centers.append(new_center)
            else:
                distances[np.argmax(distances)] = 0
                next_center = data[np.argmax(distances)]
                centers.append(next_center)
    return np.array(sorted(centers))

def apply_kmeans_to_histogram(histogram, k=3):
    """Apply K-means clustering using custom initialization."""
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot proceed with K-means clustering.")
    
    pixel_values = np.arange(len(histogram))
    probabilities = histogram / histogram_sum
    sample_size = min(10000, int(histogram_sum))
    
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)
    initial_centers = initialize_kmeans_centers(histogram, k)
    
    kmeans = KMeans(n_clusters=k, init=initial_centers.reshape(-1, 1), n_init=1, random_state=42)
    kmeans.fit(data.reshape(-1, 1))

    centers = np.sort(kmeans.cluster_centers_.flatten())
    thresholds = [(centers[i] + centers[i + 1]) / 2 for i in range(len(centers) - 1)]
    return thresholds
def process_images_and_save_results(images, histogram_save_path, csv_file_path):
    os.makedirs(histogram_save_path, exist_ok=True)  # Ensure the directory exists
    
    # Initialize sample volume and other global variables
    sample_volume = 0
    total_void_volume = 0
    total_particle_volume = 0  # Changed from fiber to particle for consistency
    total_matrix_volume = 0

    # Open a CSV file to write the thresholds
    with open(csv_file_path, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Image', 'Threshold 1', 'Threshold 2'])  # CSV headers
        
        for idx, (filename, image) in enumerate(images):
            # Convert to grayscale
            gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            
            # Calculate histogram and flatten it
            histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
            
            # Apply K-means clustering to detect thresholds
            thresholds = apply_kmeans_to_histogram(histogram)
            
            # Plot and save the histogram
            plt.figure()
            plt.title(f'Grayscale Histogram for {filename}')
            plt.xlabel('Pixel Values')
            plt.ylabel('Frequency')
            plt.plot(histogram)
            plt.xlim([0, 256])
            plt.axvline(x=thresholds[0], color='red', linestyle='--', label=f'T1: {thresholds[0]:.2f}')
            plt.axvline(x=thresholds[1], color='blue', linestyle='--', label=f'T2: {thresholds[1]:.2f}')
            plt.legend()
            
            # Save the histogram plot
            histogram_filename = f"histogram_{idx + 1}.png"
            plt.savefig(os.path.join(histogram_save_path, histogram_filename))
            plt.close()  # Close the plot to free memory
            
            # Write the filename and thresholds to the CSV file
            writer.writerow([filename, thresholds[0], thresholds[1]])
            print(f"Processed {filename}: Thresholds -> {thresholds}")
            
            # Calculate slice volume : Results in microns
            slice_volume = (gray_image.shape[0] * gray_image.shape[1] * (MICRONS_PER_PIXEL ** 2) * SLICE_THICKNESS_MICRONS)
            sample_volume += slice_volume
            
            # Void detection
            T1 = thresholds[0]
            _, binary_img = cv2.threshold(gray_image, int(T1), 255, cv2.THRESH_BINARY_INV) # Inverse is applied, voids are now white
            num_labels, labels_im = cv2.connectedComponents(binary_img)
            #All connected components and the background are labelled, labelled - 1 = number of voids (one is for the background)
            small_void_areas = 0
            void_areas = []  # List to store large void areas in pixels

            
            for label in range(1, num_labels):  # Ignore label 0 (background)
                region_mask = (labels_im == label)
                region_area = np.sum(region_mask)  # Region area in pixels

                if region_area <= MAX_VOID_AREA_PIXELS:
                    small_void_areas += region_area  # Add small void areas to cumulative sum
                else:
                    void_areas.append(region_area)  # Append large void areas to list
            
            # Convert void areas to microns
            small_void_area_microns = small_void_areas * (MICRONS_PER_PIXEL ** 2) # gets added to total particle area
            void_areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in void_areas]
            
            # Particle detection (changed from fiber to particle for consistency)
            T2 = thresholds[1]
            # Fixed potential precision issue by using T2 directly
            _, binary_particle_img = cv2.threshold(gray_image, int(T2), 255, cv2.THRESH_BINARY)

            # Find contours for particles
            num_labels, labels_particle = cv2.connectedComponents(binary_particle_img)

            particle_area = 0  # Initialize particle area for this image
            for label in range(1, num_labels):  # Iterate over each particle component
                particle_mask = (labels_particle == label)
                particle_area += np.sum(particle_mask)  # Sum the area of all particles
               
            # Void volume calculation : Results in microns
            total_void_area_microns = sum(void_areas_microns) if void_areas_microns else 0
            void_volume_microns = total_void_area_microns * SLICE_THICKNESS_MICRONS
            total_void_volume += void_volume_microns
            
            # Particle volume calculation
            particle_area_microns = particle_area * (MICRONS_PER_PIXEL ** 2) # Convert to microns
            total_particle_area_microns = particle_area_microns + small_void_area_microns
            total_particle_volume_microns = total_particle_area_microns * SLICE_THICKNESS_MICRONS
            total_particle_volume += total_particle_volume_microns

            # Create mask for matrix region (pixel intensity between T1 and T2)
            matrix_mask = (gray_image > T1) & (gray_image <= T2)

            # Calculate the number of pixels in the matrix region
            num_matrix_pixels = np.sum(matrix_mask)

            # Convert pixel count to area and then to volume
            matrix_area = num_matrix_pixels * (MICRONS_PER_PIXEL ** 2)  # Area in microns²
            matrix_volume = matrix_area * SLICE_THICKNESS_MICRONS  # Volume in microns³
            total_matrix_volume += matrix_volume
        # Calculate volume fractions with checks for division by zero
        if sample_volume == 0:
            raise ValueError("Sample volume is zero. Check image dimensions or input data.")
        else:
            void_volume_fraction = (total_void_volume / sample_volume) * 100
            particle_volume_fraction = (total_particle_volume / sample_volume) * 100
            matrix_volume_fraction = (total_matrix_volume / sample_volume) * 100
        
        print(f"Total Sample Volume: {sample_volume:.2f} µm³")
        print(f"Total Void Volume: {total_void_volume:.2f} µm³")
        print(f"Total Particle Volume: {total_particle_volume:.2f} µm³")  # Changed from fiber to particle
        print(f"Void Volume Fraction: {void_volume_fraction:.4f}%")
        print(f"Particle Volume Fraction: {particle_volume_fraction:.4f}%")  # Changed from fiber to particle
        print(f"Matrix Volume Fraction: {matrix_volume_fraction:.4f}%")

# Function to load threshold values from CSV
def load_thresholds_from_csv(csv_file_path):
    """
    Load thresholds from the given CSV file.
    Assumes the CSV file contains 'Image', 'Threshold 1', and 'Threshold 2' columns.
    Finds the largest T1 and smallest T2 for global threshold recommendation.
    """
    try:
        # Load the CSV file into a DataFrame
        thresholds_df = pd.read_csv(csv_file_path, encoding='latin1')

        # Convert to dictionary format { 'image_name': (T1, T2) }
        thresholds_dict = {row['Image']: (row['Threshold 1'], row['Threshold 2']) for _, row in thresholds_df.iterrows()}

        # Extract T1 and T2 values
        t1_values = thresholds_df["Threshold 1"].tolist()
        t2_values = thresholds_df["Threshold 2"].tolist()

        if not t1_values or not t2_values:
            raise ValueError("No valid threshold values found in the CSV.")

        # Find largest T1 and smallest T2
        global_t1 = max(t1_values)  # Largest T1
        global_t2 = min(t2_values)  # Smallest T2

        # Print recommended global threshold
        print("\n*** Recommended Global Threshold ***")
        print(f"T1: {global_t1} (Largest Threshold Value)")
        print(f"T2: {global_t2} (Smallest Threshold Value)\n")

        print("Thresholds loaded successfully from CSV.")
        return thresholds_dict, global_t1, global_t2

    except Exception as e:
        print(f"Error loading CSV file: {e}")
        return None, None, None

# Function to segment voids using threshold T1
def segment_voids_with_t1(image, t1_threshold):
    """Segment voids using the T1 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t1_threshold), 255, cv2.THRESH_BINARY_INV)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to segment particles using threshold T2
def segment_particles_with_t2(image, t2_threshold):
    """Segment particles using the T2 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t2_threshold), 255, cv2.THRESH_BINARY)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to classify voids based on shape and size
def classify_voids(binary_mask, circularity_threshold=0.87, aspect_ratio_threshold=1.5):
    """
    Classify voids into raster voids, spherical voids, and large voids based on shape parameters.
    
    Returns:
    - raster_voids: List of contours for elongated, non-circular voids
    - spherical_voids: List of contours for circular, small voids
    - large_voids: List of contours for voids larger than the size threshold
    """
    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    raster_voids = []
    spherical_voids = []
    large_voids = []

    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Ignore very small voids
            continue

        perimeter = cv2.arcLength(contour, True)
        # Added check for perimeter to avoid division by zero
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensures min(w, h) is at least 1

        # Calculate size in micrometers
        void_diameter = np.sqrt(4 * area / np.pi) * MICRONS_PER_PIXEL  # Convert pixel area to diameter

        # If the void is large (>50µm), classify separately
        if void_diameter > MAX_VOID_SIZE_MICRONS:
            large_voids.append(contour)
        elif circularity > circularity_threshold and aspect_ratio < aspect_ratio_threshold:
            spherical_voids.append(contour)
        else:
            raster_voids.append(contour)

    return raster_voids, spherical_voids, large_voids

def extract_nnd_metrics(void_mask):
    """
    Extract nearest neighbor distance (NND) metrics from a binary void mask.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing NND metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Label connected components
    labels = measure.label(binary_mask)
    regions = measure.regionprops(labels)
    
    # Need at least 2 voids for NND analysis
    if len(regions) < 2:
        return {
            'void_count': len(regions),
            'mean_nnd': None,
            'expected_nnd': None,
            'r_index': None
        }
    
    # Extract centroids
    centroids = np.array([r.centroid for r in regions])
    
    # Build KD-tree and query for nearest neighbors
    tree = cKDTree(centroids)
    distances, _ = tree.query(centroids, k=2)
    nnd = distances[:, 1]  # First column is distance to self (0), second is to nearest neighbor
    
    # Calculate spatial statistics
    area = void_mask.shape[0] * void_mask.shape[1]
    lambda_density = len(centroids) / area
    expected_nnd = 1 / (2 * np.sqrt(lambda_density))
    observed_nnd = np.mean(nnd)
    r_index = observed_nnd / expected_nnd if expected_nnd > 0 else None
    
    return {
        'void_count': len(regions),
        'mean_nnd': observed_nnd,
        'expected_nnd': expected_nnd,
        'r_index': r_index,
        'nnd_values': nnd.tolist(),  # Store all NND values for further analysis
        'nnd_std': np.std(nnd),
        'nnd_min': np.min(nnd),
        'nnd_max': np.max(nnd),
        'nnd_cv': np.std(nnd) / observed_nnd if observed_nnd > 0 else None  # Coefficient of variation
    }

# Add this function after the classify_voids function
def extract_void_morphology_metrics(void_mask):
    """
    Extract comprehensive void morphology, distribution, and orientation metrics.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing void metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Find contours for voids
    contours, _ = cv2.findContours(binary_mask.astype(np.uint8) * 255, 
                                cv2.RETR_EXTERNAL, 
                                cv2.CHAIN_APPROX_SIMPLE)
    
    # Skip if no contours found
    if not contours:
        return None
    
    # Initialize lists for metrics
    circularity_values = []
    aspect_ratio_values = []
    areas = []
    orientations = []
    centers = []
    
    # Process each void
    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Skip very small voids
            continue
            
        perimeter = cv2.arcLength(contour, True)
        # Calculate circularity with safety check for perimeter
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        # Get bounding rectangle for aspect ratio and orientation
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensure denominator is at least 1
        
        # Calculate orientation based on width/height
        orientation_rad = np.arctan2(h, w)
        orientation_deg = orientation_rad * 180 / np.pi
        
        # Get void center
        M = cv2.moments(contour)
        if M["m00"] > 0:
            center_x = int(M["m10"] / M["m00"])
            center_y = int(M["m01"] / M["m00"])
            centers.append((center_x, center_y))
        
        # Store calculated values
        circularity_values.append(circularity)
        aspect_ratio_values.append(aspect_ratio)
        areas.append(area)
        orientations.append(orientation_deg)
    
    # Skip further processing if too few voids
    if len(centers) < 2:
        return {
            'void_count': len(areas),
            'circularity_mean': np.mean(circularity_values) if circularity_values else 0,
            'aspect_ratio_mean': np.mean(aspect_ratio_values) if aspect_ratio_values else 0,
            'area_mean': np.mean(areas) if areas else 0,
        }
    
    # Calculate nearest neighbor distances
    centers_array = np.array(centers)
    tree = cKDTree(centers_array)
    distances, _ = tree.query(centers_array, k=2)  # k=2 because first result is the point itself
    nn_distances = distances[:, 1]  # Take second column (first neighbor)
    
    # Calculate clustering factor (coefficient of variation of nn distances)
    clustering_factor = np.std(nn_distances) / np.mean(nn_distances) if np.mean(nn_distances) > 0 else 0
    
    # Calculate orientation distribution
    orient_0_30_pct = ((np.array(orientations) >= 0) & (np.array(orientations) < 30)).mean() * 100
    orient_30_60_pct = ((np.array(orientations) >= 30) & (np.array(orientations) < 60)).mean() * 100
    orient_60_90_pct = ((np.array(orientations) >= 60) & (np.array(orientations) <= 90)).mean() * 100
    
    # Convert areas to physical units
    areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in areas]
    
    # Calculate orientation tensor (2D simplified version)
    nx_values = np.cos(np.array(orientations) * np.pi / 180)
    ny_values = np.sin(np.array(orientations) * np.pi / 180)
    
    orientation_tensor = np.zeros((2, 2))
    for nx, ny in zip(nx_values, ny_values):
        n = np.array([nx, ny])
        orientation_tensor += np.outer(n, n)
    
    # Normalize tensor
    orientation_tensor /= len(nx_values)
    
    # Get eigenvalues for alignment index
    eigenvalues, _ = np.linalg.eigh(orientation_tensor)
    alignment_index = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0]) if (eigenvalues[1] + eigenvalues[0]) > 0 else 0
    
    # Compile morphology metrics
    morphology_metrics = {
        # Basic void counts and statistics
        'void_count': len(areas),
        
        # Morphology metrics
        'circularity_mean': np.mean(circularity_values),
        'circularity_std': np.std(circularity_values),
        'aspect_ratio_mean': np.mean(aspect_ratio_values),
        'aspect_ratio_std': np.std(aspect_ratio_values),
        'elongated_void_percentage': (np.array(aspect_ratio_values) > 2).mean() * 100,
        'circular_void_percentage': (np.array(circularity_values) > 0.8).mean() * 100,
        
        # Distribution metrics
        'void_area_mean': np.mean(areas_microns),
        'void_area_skewness': stats.skew(areas_microns),
        'void_nn_distance': np.mean(nn_distances),
        'void_clustering': clustering_factor,
        
        # Orientation metrics
        'orientation_0_30_pct': orient_0_30_pct,
        'orientation_30_60_pct': orient_30_60_pct,
        'orientation_60_90_pct': orient_60_90_pct,
        'orientation_alignment': alignment_index
    }
    
    return morphology_metrics

# Function to annotate void segmentation
def annotate_voids(image, binary_mask):
    """
    Annotate the segmented voids on the original image.
    
    Returns:
    - Annotated image
    - Count of each void type
    """
    img_color = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
    raster_voids, spherical_voids, large_voids = classify_voids(binary_mask)

    # Draw raster voids with thinner contours (Green)
    cv2.drawContours(img_color, raster_voids, -1, (0, 255, 0), 1)  # Green for raster voids

    # Draw spherical voids (Red, thinner)
    for contour in spherical_voids:
        cv2.drawContours(img_color, [contour], -1, (0, 0, 255), 1)  # Red contour
        (x, y), radius = cv2.minEnclosingCircle(contour)
        cv2.circle(img_color, (int(x), int(y)), int(radius), (0, 0, 255), 1)  # Red enclosing circle

    # Draw large voids separately (Thicker Red Contours)
    cv2.drawContours(img_color, large_voids, -1, (0, 0, 255), 2)  # Thicker red for large voids

    return img_color, len(raster_voids), len(spherical_voids), len(large_voids)

# Function to create 3D volume from segmented slices
def create_3d_volume(segmented_slices, slice_thickness=SLICE_THICKNESS_MICRONS):
    """
    Create a 3D volume array from a series of 2D segmented slices.
    
    Parameters:
    - segmented_slices: List of binary images where features are marked as 1 or 255
    - slice_thickness: Thickness of each slice in microns
    
    Returns:
    - 3D numpy array representing the stacked volume
    """
    # Check if we have any slices
    if not segmented_slices:
        raise ValueError("No slices provided for 3D volume creation")
    
    # Ensure all slices have the same dimensions
    heights = [slice.shape[0] for slice in segmented_slices]
    widths = [slice.shape[1] for slice in segmented_slices]
    
    if len(set(heights)) > 1 or len(set(widths)) > 1:
        # Resize slices if dimensions vary
        h, w = max(heights), max(widths)
        for i, slice in enumerate(segmented_slices):
            if slice.shape[0] != h or slice.shape[1] != w:
                segmented_slices[i] = cv2.resize(slice, (w, h))
    
    # Normalize slices to binary (0, 1) values
    normalized_slices = []
    for slice in segmented_slices:
        if slice.max() > 1:
            normalized_slices.append(slice > 0)  # Convert to boolean/binary
        else:
            normalized_slices.append(slice)
    
    # Stack slices to create 3D volume
    volume = np.stack(normalized_slices, axis=0)
    
    return volume

def create_mesh_surface(volume, voxel_size):
    """Helper function to create a PyVista mesh from a binary volume"""
    try:
        # Check if volume has any positive values
        if not np.any(volume):
            print("Warning: Volume is empty. Cannot create mesh.")
            return None
        
        # Extract surface using marching cubes
        verts, faces, normals, values = measure.marching_cubes(volume, 0.5)
        
        # Adjust vertices by voxel size - fix dimension ordering
        verts[:, 0] *= voxel_size[0]  # X dimension
        verts[:, 1] *= voxel_size[1]  # Y dimension
        verts[:, 2] *= voxel_size[2]  # Z dimension
        
        # Create PyVista mesh
        faces = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        mesh = pv.PolyData(verts, faces)
        
        # Smooth the mesh for better visualization
        mesh = mesh.smooth(n_iter=15)
        
        return mesh
    except Exception as e:
        print(f"Error creating mesh surface: {str(e)}")
        return None

def create_professional_3d_visualization(void_volume, particle_volume, output_path, 
                                        voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """Create a high-quality 3D visualization of voids, particles, and matrix."""
    try:
        # Suppress warnings
        warnings.filterwarnings("ignore", category=UserWarning)
        warnings.filterwarnings("ignore", category=DeprecationWarning)

        # Check for valid data
        if not np.any(void_volume) and not np.any(particle_volume):
            raise ValueError("Both void and particle volumes are empty.")

        # Configure plotter for compatibility
        pv.set_plot_theme(pv.themes.DocumentTheme())
        plotter = pv.Plotter(
            off_screen=True,
            window_size=[1200, 1000],
            lighting='three lights',
            polygon_smoothing=False
        )

        # Convert voxel size to mm
        voxel_size_mm = (voxel_size[0]/1000, voxel_size[1]/1000, voxel_size[2]/1000)

        # Create and add meshes
        matrix_volume = ~(void_volume | particle_volume)
        
        components = [
            (matrix_volume, 'tan', 'Matrix', 0.6),
            (void_volume, 'blue', 'Voids', 0.4),
            (particle_volume, 'darkred', 'Particles', 0.8)
        ]

        for vol, color, name, opacity in components:
            if np.any(vol):
                mesh = create_mesh_surface(vol, voxel_size)
                if mesh:
                    plotter.add_mesh(mesh, color=color, opacity=opacity,
                                   smooth_shading=False, specular=0.5)

        # Configure axes and labels
        plotter.show_bounds(
            xtitle='X (mm)', 
            ytitle='Y (mm)', 
            ztitle='Z (mm)',
            grid=True,
            location='outer',
            all_edges=True
        )

        # Add legend and title
        plotter.add_legend(
            [('Matrix', 'tan'), ('Voids', 'blue'), ('Particles', 'darkred')],
            size=(0.15, 0.15),
            loc='upper right'
        )
        plotter.add_text("HDPE-HGM Syntactic Foam - 3D Reconstruction", 
                        position='upper_left', font_size=14)

        # Set camera and save output
        plotter.view_isometric()
        plotter.camera.zoom(0.75)
        
        # Create output directory if needed
        output_dir = os.path.dirname(output_path)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)

        # Handle different output formats
        file_ext = os.path.splitext(output_path)[1].lower()
        
        if file_ext == '.html':
            plotter.export_html(output_path)
            html_path = output_path
        else:
            plotter.screenshot(output_path)
            # Create companion HTML file
            html_path = os.path.splitext(output_path)[0] + '.html'
            plotter.export_html(html_path)

        plotter.close()
        print(f"Visualization saved to: {output_path}")
        return output_path, html_path  # Return tuple consistently

    except Exception as e:
        print(f"Visualization error: {str(e)}")
        return None, None  # Return tuple for consistency
        
def generate_mesh_from_volume(volume, output_file, voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """
    Generate a mesh file from a 3D binary volume.
    
    Parameters:
    - volume: 3D binary array
    - output_file: Path to save the mesh file
    - voxel_size: Tuple of (x, y, z) dimensions in microns
    
    Returns:
    - mesh: PyVista mesh object
    """
    # Ensure the volume is binary
    binary_volume = volume > 0
    
    # Check if volume has any positive values before processing
    if not np.any(binary_volume):
        print(f"Warning: Volume for {output_file} is empty. Cannot generate mesh.")
        return None
    
    # Extract surface mesh using marching cubes
    try:
        verts, faces, normals, values = measure.marching_cubes(binary_volume, 0.5)
        
        # Adjust vertices by voxel size - corrected dimension ordering
        verts[:, 0] *= voxel_size[0]
        verts[:, 1] *= voxel_size[1]
        verts[:, 2] *= voxel_size[2]
        
        # Convert faces to the format expected by PyVista
        faces_pv = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        
        # Create PyVista mesh
        mesh = pv.PolyData(verts, faces_pv)
        
        # Make sure the output directory exists
        output_dir = os.path.dirname(output_file)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)
        
        # Save mesh to file
        file_ext = os.path.splitext(output_file)[1].lower()
        
        if file_ext == '.stl':
            mesh.save(output_file, binary=True)
        elif file_ext == '.obj':
            mesh.save(output_file)
        elif file_ext == '.ply':
            mesh.save(output_file)
        else:
            # Default to STL if extension is not recognized
            output_file = os.path.splitext(output_file)[0] + '.stl'
            mesh.save(output_file, binary=True)
        
        print(f"Mesh saved to {output_file}")
        return mesh
        
    except Exception as e:
        print(f"Error generating mesh: {str(e)}")
        return None

# Function to generate separate meshes for voids and particles
def generate_component_meshes(void_volume, particle_volume, output_dir):
    """
    Generate separate mesh files for voids and particles.
    
    Parameters:
    - void_volume: 3D array with voids marked as 1
    - particle_volume: 3D array with particles marked as 1
    - output_dir: Directory to save mesh files
    
    Returns:
    - Dictionary with mesh file paths
    """
    os.makedirs(output_dir, exist_ok=True)
    
    mesh_files = {}
    
    # Generate void mesh
    void_mesh_path = os.path.join(output_dir, "voids_mesh.stl")
    void_mesh = generate_mesh_from_volume(void_volume, void_mesh_path)
    if void_mesh:
        mesh_files['voids'] = void_mesh_path
    
    # Generate particle mesh
    particle_mesh_path = os.path.join(output_dir, "particles_mesh.stl")
    particle_mesh = generate_mesh_from_volume(particle_volume, particle_mesh_path)
    if particle_mesh:
        mesh_files['particles'] = particle_mesh_path
          # Generate combined sample mesh (matrix material)
    matrix_volume = ~(void_volume | particle_volume)
    matrix_mesh_path = os.path.join(output_dir, "matrix_mesh.stl")
    matrix_mesh = generate_mesh_from_volume(matrix_volume, matrix_mesh_path)
    if matrix_mesh:
        mesh_files['matrix'] = matrix_mesh_path
    
    # Generate complete sample mesh (all components together)
    complete_mesh_path = os.path.join(output_dir, "complete_sample.vtm")
    
    # Create multiple meshes in a single VTK file
    blocks = pv.MultiBlock()
    
    if void_mesh:
        blocks.append(void_mesh)
    if particle_mesh:
        blocks.append(particle_mesh)
    if matrix_mesh:
        blocks.append(matrix_mesh)
    
    # Only save if we have at least one valid mesh
    if len(blocks) > 0:
        blocks.save(complete_mesh_path)
        mesh_files['complete'] = complete_mesh_path
    
    return mesh_files

# Function to process a stack of images and create 3D visualizations and meshes
def process_image_stack(images, thresholds_dict, output_dir):
    """
    Process a stack of images to create 3D visualizations and meshes.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Base directory for outputs
    
    Returns:
    - Dictionary with results and file paths
    """
    # Create output subdirectories
    visualization_dir = os.path.join(output_dir, "3d_visualizations")
    mesh_dir = os.path.join(output_dir, "meshes")
    segmentation_dir = os.path.join(output_dir, "segmented_images")
    morphology_dir = os.path.join(output_dir, "void_morphology")
    
    for directory in [visualization_dir, mesh_dir, segmentation_dir, morphology_dir]:
        os.makedirs(directory, exist_ok=True)
    
    void_slices = []
    particle_slices = []
    segmented_images = []
    
    raster_void_count = 0
    spherical_void_count = 0
    large_void_count = 0
    
    # New addition: Create lists to store void metrics for each slice
    void_morphology_metrics_per_slice = []
    void_details_list = []
    
    # Process each image
    for idx, (filename, img) in enumerate(images):
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            t1, t2 = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment voids and particles
        void_mask = segment_voids_with_t1(gray_img, t1)
        particle_mask = segment_particles_with_t2(gray_img, t2)
        
        # Annotate voids
        annotated_img, raster_count, spherical_count, large_count = annotate_voids(gray_img, void_mask)
        
        # NEW: Extract NND metrics
        nnd_metrics = extract_nnd_metrics(void_mask)
        if nnd_metrics:
            nnd_metrics['filename'] = filename
            nnd_metrics['slice_index'] = idx
            void_morphology_metrics_per_slice.append(nnd_metrics)
        
        # NEW: Extract void details for each individual void
        labels = measure.label(void_mask > 0)
        regions = measure.regionprops(labels)
        
        for i, region in enumerate(regions):
            y, x = region.centroid
            width = region.bbox[3] - region.bbox[1]
            height = region.bbox[2] - region.bbox[0]
            area = region.area
            perimeter = region.perimeter if hasattr(region, 'perimeter') else 0
            
            # Calculate circularity and aspect ratio
            circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
            aspect_ratio = max(width, height) / max(min(width, height), 1)
            
            # Calculate orientation
            orientation_rad = np.arctan2(height, width)
            orientation_deg = orientation_rad * 180 / np.pi
            
            # Store void details
            void_details = {
                'Slice': idx + 1,
                'Image': filename,
                'ID': i + 1,
                'X': x,
                'Y': y,
                'Width': width,
                'Height': height,
                'Area': area,
                'Circularity': circularity,
                'Aspect Ratio': aspect_ratio,
                'Orientation_deg': orientation_deg
            }
            void_details_list.append(void_details)
        
        # Update counts
        raster_void_count += raster_count
        spherical_void_count += spherical_count
        large_void_count += large_count
        
        # Save segmented image
        segmented_path = os.path.join(segmentation_dir, f"segmented_{filename}")
        cv2.imwrite(segmented_path, annotated_img)
        segmented_images.append(segmented_path)
        
        # Add segmentation masks to slices for 3D reconstruction
        void_slices.append(void_mask)
        particle_slices.append(particle_mask)
    
    # NEW: Save void details to CSV
    if void_details_list:
        void_details_df = pd.DataFrame(void_details_list)
        void_details_csv_path = os.path.join(output_dir, "Void_Details.csv")
        void_details_df.to_csv(void_details_csv_path, index=False)
        print(f"Void details saved to: {void_details_csv_path}")
    
    # Check if we have any valid slices
    if not void_slices or not particle_slices:
        raise ValueError("No valid slices found for 3D reconstruction")
    
    # Create 3D volumes
    void_volume = create_3d_volume(void_slices)
    particle_volume = create_3d_volume(particle_slices)
    
    # Create 3D visualization
    viz_path = os.path.join(visualization_dir, "3d_segmentation.png")
    viz_path, html_path = create_professional_3d_visualization(
        void_volume, particle_volume, viz_path
    )
    
    # Generate mesh files
    mesh_files = generate_component_meshes(void_volume, particle_volume, mesh_dir)
    
    # Calculate volume fractions with safe division
    total_volume = void_volume.size
    if total_volume > 0:
        void_volume_fraction = (np.sum(void_volume) / total_volume) * 100
        particle_volume_fraction = (np.sum(particle_volume) / total_volume) * 100
        matrix_volume_fraction = 100 - void_volume_fraction - particle_volume_fraction
    else:
        void_volume_fraction = 0
        particle_volume_fraction = 0
        matrix_volume_fraction = 0
    
    # NEW: After processing all slices, aggregate morphology metrics
    aggregate_void_metrics = {}
    
    if void_morphology_metrics_per_slice:
        # Calculate mean values across all slices
        for key in void_morphology_metrics_per_slice[0].keys():
            if key not in ['filename', 'slice_index', 'nnd_values'] and isinstance(void_morphology_metrics_per_slice[0][key], (int, float)) and void_morphology_metrics_per_slice[0][key] is not None:
                values = [metrics[key] for metrics in void_morphology_metrics_per_slice if key in metrics and metrics[key] is not None]
                if values:
                    aggregate_void_metrics[key] = np.mean(values)
    
    # NEW: Calculate global morphology statistics from void details
    if void_details_list:
        void_df = pd.DataFrame(void_details_list)
        morphology_stats = {
            'circularity_mean': void_df['Circularity'].mean(),
            'circularity_std': void_df['Circularity'].std(),
            'aspect_ratio_mean': void_df['Aspect Ratio'].mean(),
            'aspect_ratio_std': void_df['Aspect Ratio'].std(),
            'elongated_void_percentage': (void_df['Aspect Ratio'] > 2).mean() * 100,
            'circular_void_percentage': (void_df['Circularity'] > 0.8).mean() * 100,
            'void_area_mean': void_df['Area'].mean(),
            'void_area_skewness': void_df['Area'].skew()
        }
        
        # Calculate orientation distribution
        if 'Orientation_deg' in void_df.columns:
            morphology_stats['orientation_0_30_pct'] = ((void_df['Orientation_deg'] >= 0) & 
                                                      (void_df['Orientation_deg'] < 30)).mean() * 100
            morphology_stats['orientation_30_60_pct'] = ((void_df['Orientation_deg'] >= 30) & 
                                                        (void_df['Orientation_deg'] < 60)).mean() * 100
            morphology_stats['orientation_60_90_pct'] = ((void_df['Orientation_deg'] >= 60) & 
                                                        (void_df['Orientation_deg'] <= 90)).mean() * 100
            
            # Calculate orientation tensor
            rad_angles = void_df['Orientation_deg'] * np.pi / 180
            nx_values = np.cos(rad_angles)
            ny_values = np.sin(rad_angles)
            
            orientation_tensor = np.zeros((2, 2))
            for nx, ny in zip(nx_values, ny_values):
                n = np.array([nx, ny])
                orientation_tensor += np.outer(n, n)
            
            if len(nx_values) > 0:
                orientation_tensor /= len(nx_values)
                eigenvalues, _ = np.linalg.eigh(orientation_tensor)
                if (eigenvalues[1] + eigenvalues[0]) > 0:
                    morphology_stats['orientation_alignment'] = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0])
        
        # Update aggregate metrics with morphology stats
        aggregate_void_metrics.update(morphology_stats)
    
    # Compile results
    results = {
        "raster_void_count": raster_void_count,
        "spherical_void_count": spherical_void_count,
        "large_void_count": large_void_count,
        "total_void_count": raster_void_count + spherical_void_count + large_void_count,
        "void_volume_fraction": void_volume_fraction,
        "particle_volume_fraction": particle_volume_fraction,
        "matrix_volume_fraction": matrix_volume_fraction,
        "visualization_path": viz_path,
        "html_visualization_path": html_path,
        "mesh_files": mesh_files,
        "segmented_images": segmented_images,
        "void_volume_shape": void_volume.shape,
        "physical_dimensions_mm": (
            void_volume.shape[2] * MICRONS_PER_PIXEL / 1000,
            void_volume.shape[1] * MICRONS_PER_PIXEL / 1000,
            void_volume.shape[0] * SLICE_THICKNESS_MICRONS / 1000
        ),
        # NEW: Add the void morphology metrics
        "void_morphology_metrics": aggregate_void_metrics
    }
    
    # Save results to CSV
    results_csv_path = os.path.join(output_dir, "3d_analysis_results.csv")
    with open(results_csv_path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['Parameter', 'Value', 'Units'])
        writer.writerow(['Raster Void Count', raster_void_count, ''])
        writer.writerow(['Spherical Void Count', spherical_void_count, ''])
        writer.writerow(['Large Void Count', large_void_count, ''])
        writer.writerow(['Total Void Count', results["total_void_count"], ''])
        writer.writerow(['Void Volume Fraction', f"{void_volume_fraction:.4f}", '%'])
        writer.writerow(['Particle Volume Fraction', f"{particle_volume_fraction:.4f}", '%'])
        writer.writerow(['Matrix Volume Fraction', f"{matrix_volume_fraction:.4f}", '%'])
        writer.writerow(['X Dimension', f"{results['physical_dimensions_mm'][0]:.4f}", 'mm'])
        writer.writerow(['Y Dimension', f"{results['physical_dimensions_mm'][1]:.4f}", 'mm'])
        writer.writerow(['Z Dimension', f"{results['physical_dimensions_mm'][2]:.4f}", 'mm'])
        
        # NEW: Add void morphology metrics to CSV
        writer.writerow(['', '', ''])  # Empty row as separator
        writer.writerow(['Void Morphology and Distribution Metrics', '', ''])
        for key, value in aggregate_void_metrics.items():
            # Format the key for better readability
            formatted_key = ' '.join(word.capitalize() for word in key.split('_'))
            if isinstance(value, (int, float)):
                writer.writerow([formatted_key, f"{value:.4f}", ''])
    
    # NEW: Save detailed morphology metrics to CSV
    if void_morphology_metrics_per_slice:
        morphology_df = pd.DataFrame(void_morphology_metrics_per_slice)
        # Remove the nnd_values column which can be very large
        if 'nnd_values' in morphology_df.columns:
            morphology_df = morphology_df.drop(columns=['nnd_values'])
        morphology_csv_path = os.path.join(morphology_dir, "void_morphology_metrics.csv")
        morphology_df.to_csv(morphology_csv_path, index=False)
        print(f"Void morphology metrics saved to: {morphology_csv_path}")
    
    # NEW: Create visualization of void orientation distribution
    if void_details_list:
        void_df = pd.DataFrame(void_details_list)
        if 'Orientation_deg' in void_df.columns:
            # Create orientation rose diagram
            fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})
            
            # Convert to radians and bin the data
            orientations_rad = void_df['Orientation_deg'] * np.pi / 180
            bins = np.linspace(0, np.pi, 19)  # 18 bins covering 0 to 180 degrees
            
            # Create histogram
            hist, bin_edges = np.histogram(orientations_rad, bins=bins)
            
            # Plot rose diagram
            width = np.pi / 18
            bars = ax.bar(bin_edges[:-1], hist, width=width, alpha=0.7)
            
            # Customize appearance
            ax.set_theta_zero_location('N')  # Set 0 degrees at the top
            ax.set_theta_direction(-1)  # Clockwise
            ax.set_xticks(np.linspace(0, np.pi, 7))
            ax.set_xticklabels(['0°', '30°', '60°', '90°', '120°', '150°', '180°'])
            ax.set_title('Void Orientation Distribution', fontsize=16)
            ax.grid(True)
            
            # Save the figure
            orientation_viz_path = os.path.join(morphology_dir, 'void_orientation_distribution.png')
            plt.savefig(orientation_viz_path, dpi=300, bbox_inches='tight')
            plt.close(fig)
            print(f"Void orientation distribution visualization saved to: {orientation_viz_path}")
    
    return results

def generate_raster_void_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing raster voids
    - Dictionary of raster void details
    """
    raster_void_slices = []
    raster_void_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            t1, _ = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment voids using T1 threshold
        void_mask = segment_voids_with_t1(gray_img, t1)
        
        # Classify voids to get specifically raster voids
        raster_voids, _, _ = classify_voids(void_mask)
        
        # Create a binary mask for raster voids
        raster_void_mask = np.zeros_like(gray_img)
        cv2.drawContours(raster_void_mask, raster_voids, -1, 255, -1)
        
        # Store slice and details
        raster_void_slices.append(raster_void_mask)
        
        # Collect void details
        void_details = {
            'filename': filename,
            'raster_void_count': len(raster_voids),
            'raster_void_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in raster_voids]
        }
        raster_void_details.append(void_details)
    
    # Create 3D volume
    raster_void_volume = create_3d_volume(raster_void_slices)
    
    return raster_void_volume, raster_void_details

def generate_particle_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing particles
    - Dictionary of particle details
    """
    particle_slices = []
    particle_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            _, t2 = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment particles using T2 threshold
        particle_mask = segment_particles_with_t2(gray_img, t2)
        
        # Find particle contours
        num_labels, labels_particle = cv2.connectedComponents(particle_mask)
        
        # Create list of particle contours
        particle_contours = []
        for label in range(1, num_labels):
            component_mask = (labels_particle == label).astype(np.uint8) * 255
            contours, _ = cv2.findContours(component_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            particle_contours.extend(contours)
        
        # Create binary mask for particles
        particle_binary = np.zeros_like(gray_img)
        cv2.drawContours(particle_binary, particle_contours, -1, 255, -1)
        
        # Store slice and details
        particle_slices.append(particle_binary)
        
        # Collect particle details
        particle_details.append({
            'filename': filename,
            'particle_count': len(particle_contours),
            'particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in particle_contours]
        })
    
    # Create 3D volume
    particle_volume = create_3d_volume(particle_slices)
    
    return particle_volume, particle_details

def generate_raster_void_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "raster_void_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Collect raster void slices
        raster_void_slices = []
        raster_void_details = []

        for filename, img in images:
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                t1, _ = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment voids using T1 threshold
            void_mask = segment_voids_with_t1(gray_img, t1)
            
            # Classify voids to get specifically raster voids
            raster_voids, _, _ = classify_voids(void_mask)
            
            # Create a binary mask for raster voids
            raster_void_mask = np.zeros_like(gray_img)
            cv2.drawContours(raster_void_mask, raster_voids, -1, 255, -1)
            
            # Store slice and details
            raster_void_slices.append(raster_void_mask)
            
            # Collect void details
            void_details = {
                'filename': filename,
                'raster_void_count': len(raster_voids),
                'raster_void_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in raster_voids]
            }
            raster_void_details.append(void_details)
        
        # Create 3D volume
        raster_void_volume = create_3d_volume(raster_void_slices)
        
        # Visualize 3D volume of raster voids
        viz_path = os.path.join(visualization_dir, "raster_voids_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            raster_void_volume, 
            np.zeros_like(raster_void_volume),  # Empty particle volume
            viz_path
        )
        
        return viz_path, raster_void_volume

    except Exception as e:
        print(f"Error generating raster void 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def generate_spherical_particle_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of spherical particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "spherical_particle_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Collect spherical particle slices
        spherical_particle_slices = []
        spherical_particle_details = []

        for filename, img in images:
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                _, t2 = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment particles using T2 threshold
            particle_mask = segment_particles_with_t2(gray_img, t2)
            
            # Classify voids to extract specifically spherical particles
            _, spherical_particles, _ = classify_voids(particle_mask)
            
            # Create a binary mask for spherical particles
            spherical_particle_mask = np.zeros_like(gray_img)
            cv2.drawContours(spherical_particle_mask, spherical_particles, -1, 255, -1)
            
            # Store slice and details
            spherical_particle_slices.append(spherical_particle_mask)
            
            # Collect particle details
            particle_details = {
                'filename': filename,
                'spherical_particle_count': len(spherical_particles),
                'spherical_particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in spherical_particles]
            }
            spherical_particle_details.append(particle_details)
        
        # Create 3D volume
        spherical_particle_volume = create_3d_volume(spherical_particle_slices)
        
        # Visualize 3D volume of spherical particles
        viz_path = os.path.join(visualization_dir, "spherical_particles_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            np.zeros_like(spherical_particle_volume),  # Empty void volume
            spherical_particle_volume, 
            viz_path
        )
        
        return viz_path, spherical_particle_volume

    except Exception as e:
        print(f"Error generating spherical particle 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def visualize_3d_volumes(raster_void_volume, particle_volume, output_dir):
    """
    Visualize and save 3D volumes for raster voids and particles.
    
    Parameters:
    - raster_void_volume: 3D numpy array of raster voids
    - particle_volume: 3D numpy array of particles
    - output_dir: Directory to save visualizations
    
    Returns:
    - Dictionary with visualization paths and details
    """
    # Create output directories
    visualization_dir = os.path.join(output_dir, "3d_component_visualizations")
    mesh_dir = os.path.join(output_dir, "component_meshes")
    
    os.makedirs(visualization_dir, exist_ok=True)
    os.makedirs(mesh_dir, exist_ok=True)
    
    # Visualization for raster voids
    raster_void_viz_path = os.path.join(visualization_dir, "raster_voids_3d.png")
    raster_void_viz_path, _ = create_professional_3d_visualization(
        raster_void_volume, 
        np.zeros_like(raster_void_volume),  # Empty particle volume
        raster_void_viz_path
    )
    
    # Visualization for particles
    particle_viz_path = os.path.join(visualization_dir, "particles_3d.png")
    particle_viz_path, _ = create_professional_3d_visualization(
        np.zeros_like(particle_volume),  # Empty void volume
        particle_volume, 
        particle_viz_path
    )
    
    # Generate mesh files
    raster_void_meshes = generate_component_meshes(
        raster_void_volume, 
        np.zeros_like(raster_void_volume), 
        os.path.join(mesh_dir, "raster_void_meshes")
    )
    
    particle_meshes = generate_component_meshes(
        np.zeros_like(particle_volume),
        particle_volume, 
        os.path.join(mesh_dir, "particle_meshes")
    )
    
    # Calculate volume fractions
    total_raster_void_volume = raster_void_volume.size
    raster_void_volume_fraction = (np.sum(raster_void_volume) / total_raster_void_volume) * 100
    
    total_particle_volume_size = particle_volume.size
    particle_volume_fraction = (np.sum(particle_volume) / total_particle_volume_size) * 100
    
    # Compile results
    results = {
        "raster_void_visualization_path": raster_void_viz_path,
        "particle_visualization_path": particle_viz_path,
        "raster_void_meshes": raster_void_meshes,
        "particle_meshes": particle_meshes,
        "raster_void_volume_fraction": raster_void_volume_fraction,
        "particle_volume_fraction": particle_volume_fraction
    }
    
    return results
if __name__ == "__main__":
    root = tk.Tk()
    app = FoamAnalyzerGUI(root)
    root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox, scrolledtext
from PIL import Image, ImageTk
import os
import pandas as pd
import cv2
import numpy as np
import matplotlib.pyplot as plt
import csv
import warnings
from sklearn.cluster import KMeans
from skimage import measure
from scipy.spatial import cKDTree
from scipy import stats
import pyvista as pv

# REMOVED THE PROBLEMATIC IMPORT - all functions are defined in this file

# Global variable
_root = None

# Constants
MICRONS_PER_PIXEL = 4.89
SLICE_THICKNESS_MICRONS = 4.89
MAX_VOID_SIZE_MICRONS = 50.0
MAX_VOID_SIZE_PIXELS = MAX_VOID_SIZE_MICRONS / MICRONS_PER_PIXEL
MAX_VOID_AREA_PIXELS = np.pi * (MAX_VOID_SIZE_PIXELS / 2) ** 2

class FoamAnalyzerGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("HDPE-HGM Syntactic Foam Analyzer")
        self.root.geometry("1600x1000")
        style = ttk.Style(root)
        style.configure('TFrame', background='#f0f4f8')
        style.configure('TLabel', font=('Times New Roman', 12), background='#f0f4f8')
        style.configure('TButton', font=('Times New Roman', 12))
        style.configure('TEntry', font=('Times New Roman', 12))

        self.annotated_images = []
        self.thresholds_dict = {}

        self.setup_ui()
    
    def setup_ui(self):
        # Top controls frame
        top = ttk.Frame(self.root)
        top.pack(fill='x', pady=5, padx=10)

        # Image folder selection
        ttk.Label(top, text="Image Folder:").grid(row=0, column=0, sticky='w', padx=5)
        self.img_entry = ttk.Entry(top, width=70)
        self.img_entry.grid(row=0, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_image).grid(row=0, column=2, padx=5)

        # CSV file selection
        ttk.Label(top, text="Threshold CSV:").grid(row=1, column=0, sticky='w', padx=5)
        self.csv_entry = ttk.Entry(top, width=70)
        self.csv_entry.grid(row=1, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_csv).grid(row=1, column=2, padx=5)

        # Process button
        ttk.Button(top, text="Process", command=self.process_images).grid(row=2, column=1, pady=10)

        # Progress bar
        self.progress = ttk.Progressbar(self.root, mode='indeterminate')
        self.progress.pack(fill='x', pady=5, padx=10)

        # Middle frame for results and controls
        middle_frame = ttk.Frame(self.root)
        middle_frame.pack(fill='both', expand=True, pady=10, padx=10)

        # Left panel for image viewing
        left_panel = ttk.Frame(middle_frame)
        left_panel.pack(side='left', fill='both', expand=True, padx=5)

        # Threshold labels
        threshold_frame = ttk.Frame(left_panel)
        threshold_frame.pack(fill='x', pady=5)
        
        self.t1label = ttk.Label(threshold_frame, text="Global T1: Not calculated")
        self.t1label.pack(side='left', padx=10)
        
        self.t2label = ttk.Label(threshold_frame, text="Global T2: Not calculated")
        self.t2label.pack(side='left', padx=10)

        # Image display
        image_frame = ttk.Frame(left_panel)
        image_frame.pack(fill='both', expand=True, pady=5)
        
        self.ct_img_label = ttk.Label(image_frame, text="No images loaded", 
                                     relief='sunken', borderwidth=2)
        self.ct_img_label.pack(fill='both', expand=True)

        # Image slider
        slider_frame = ttk.Frame(left_panel)
        slider_frame.pack(fill='x', pady=5)
        
        ttk.Label(slider_frame, text="Image:").pack(side='left')
        self.slider = ttk.Scale(slider_frame, from_=0, to=0, orient='horizontal',
                               command=self.update_image)
        self.slider.pack(side='left', fill='x', expand=True, padx=5)

        # Right panel for 3D visualization and controls
        right_panel = ttk.Frame(middle_frame)
        right_panel.pack(side='right', fill='both', expand=True, padx=5)

        # 3D visualization buttons
        button_frame = ttk.Frame(right_panel)
        button_frame.pack(fill='x', pady=5)
        
        ttk.Button(button_frame, text="Show Voids 3D", 
                  command=self.show_voids3d).pack(side='left', padx=5)
        ttk.Button(button_frame, text="Show Particles 3D", 
                  command=self.show_particles3d).pack(side='left', padx=5)

        # 3D display area
        self.static_3d_label = ttk.Label(right_panel, text="3D Visualization Area", 
                                        relief='sunken', borderwidth=2)
        self.static_3d_label.pack(fill='both', expand=True, pady=5)

        # Volume fraction display
        self.vflabel = ttk.Label(right_panel, 
                                text="VF% - Process images to see results",
                                font=('Times New Roman', 10))
        self.vflabel.pack(fill='x', pady=5)
        
    def browse_image(self):
        """Browse for image folder."""
        path = filedialog.askdirectory(title="Select Image Folder")
        if path:
            self.img_entry.delete(0, 'end')
            self.img_entry.insert(0, path)

    def browse_csv(self):
        """Browse for CSV file."""
        path = filedialog.askopenfilename(
            title="Select Threshold CSV File",
            filetypes=[("CSV files", "*.csv")]
        )
        if path:
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, path)

    def process_images(self):
        """Process images and create thresholds."""
        folder = self.img_entry.get()
        csv_file = self.csv_entry.get()
        
        if not folder:
            messagebox.showerror("Error", "Please select an image folder")
            return

        # If CSV missing, prompt for save location
        if not csv_file:
            csv_file = filedialog.asksaveasfilename(
                defaultextension=".csv",
                filetypes=[("CSV files", "*.csv")],
                title="Save thresholds CSV"
            )
            if not csv_file:
                messagebox.showerror("Error", "Threshold CSV file path required")
                return
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, csv_file)

        # Start progress bar
        self.progress.start()
        self.root.update()

        try:
            # Load and process images
            images = load_images_from_folder(folder)
            if not images:
                raise ValueError("No valid images found in the selected folder")
            
            # Process images and save results
            process_images_and_save_results(images, "histograms", csv_file)
            
            # Load thresholds
            threshold_result = load_thresholds_from_csv(csv_file)
            if threshold_result[0] is None:
                raise ValueError("Failed to load thresholds from CSV")
            
            self.thresholds_dict, g1, g2 = threshold_result
            self.t1label.config(text=f"Global T1: {g1:.2f}")
            self.t2label.config(text=f"Global T2: {g2:.2f}")

            # Load annotated images for display
            self.annotated_images = []
            for filename, img in images:
                # Convert to display format
                if len(img.shape) == 3:
                    display_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                else:
                    display_img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
                
                pil_img = Image.fromarray(display_img).resize((400, 400))
                self.annotated_images.append(pil_img)
            
            # Update slider and display first image
            if self.annotated_images:
                self.slider.config(to=len(self.annotated_images)-1)
                self.update_image(0)
            
            messagebox.showinfo("Success", "Image processing completed successfully!")
            
        except Exception as e:
            messagebox.showerror("Error", f"Processing failed: {str(e)}")
        finally:
            self.progress.stop()

    def update_image(self, val):
        """Update displayed image based on slider value."""
        try:
            idx = int(float(val))
            if 0 <= idx < len(self.annotated_images):
                img = self.annotated_images[idx]
                imgtk = ImageTk.PhotoImage(img)
                self.ct_img_label.config(image=imgtk, text="")
                self.ct_img_label.image = imgtk  # Keep a reference
        except (ValueError, IndexError):
            pass  # Ignore invalid values

    def show_voids3d(self):
        """Generate and display 3D void visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="")
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")

    def show_particles3d(self):
        """Generate and display 3D particle visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="Particles 3D", compound='top')
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")


# MOVED get_root() OUTSIDE THE CLASS - FIXED INDENTATION
def get_root():
    global _root
    if _root is None:
        _root = tk.Tk()
        _root.withdraw()
    return _root

# REMOVED unused select_folder and select_csv_file functions to avoid confusion

def load_images_from_folder(folder_path):
    """Load images from the specified folder."""
    images = []
    for filename in sorted(os.listdir(folder_path)):
        if filename.lower().endswith(('.tif', '.tiff', '.png', '.jpg', '.jpeg')):
            img_path = os.path.join(folder_path, filename)
            img = cv2.imread(img_path)
            if img is not None:
                images.append((filename, img))  # Append as (filename, image)
    return images

def calculate_histogram(gray_image):
    """Calculate and return the histogram of a grayscale image."""
    histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
    return histogram

def calculate_epsilon(histogram, sample_size=10000):
    """Compute epsilon for neighborhood initialization (approximate mean pairwise distance)."""
    pixel_values = np.arange(256)
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot calculate epsilon.")
    
    probabilities = histogram / histogram_sum  # Normalize histogram to get probabilities
    sample_size = int(min(sample_size, histogram_sum))

    sampled_data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    mean_distance = np.mean(np.abs(np.diff(np.sort(sampled_data))))
    return mean_distance

def initialize_kmeans_centers(histogram, k, epsilon=None):
    pixel_values = np.arange(len(histogram))
    histogram_sum = histogram.sum()

    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot initialize K-means centers.")

    probabilities = histogram / histogram_sum  # Normalize histogram
    sample_size = int(min(10000, histogram_sum))
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    if epsilon is None:
        epsilon = calculate_epsilon(histogram)

    centers = []
    for _ in range(k):
        if len(centers) == 0:
            first_center = np.argmax(histogram)
            centers.append(first_center)
        else:
            distances = np.min([np.abs(data - center) for center in centers], axis=0)
            if np.max(distances) <= epsilon:
                new_center = data[np.argmax(distances)]
            else:
                new_center = data[np.argmax(distances)]
            if new_center not in centers:
                centers.append(new_center)
            else:
                distances[np.argmax(distances)] = 0
                next_center = data[np.argmax(distances)]
                centers.append(next_center)
    return np.array(sorted(centers))

def apply_kmeans_to_histogram(histogram, k=3):
    """Apply K-means clustering using custom initialization."""
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot proceed with K-means clustering.")
    
    pixel_values = np.arange(len(histogram))
    probabilities = histogram / histogram_sum
    sample_size = min(10000, int(histogram_sum))
    
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)
    initial_centers = initialize_kmeans_centers(histogram, k)
    
    kmeans = KMeans(n_clusters=k, init=initial_centers.reshape(-1, 1), n_init=1, random_state=42)
    kmeans.fit(data.reshape(-1, 1))

    centers = np.sort(kmeans.cluster_centers_.flatten())
    thresholds = [(centers[i] + centers[i + 1]) / 2 for i in range(len(centers) - 1)]
    return thresholds

def process_images_and_save_results(images, histogram_save_path, csv_file_path):
    os.makedirs(histogram_save_path, exist_ok=True)  # Ensure the directory exists
    
    # Initialize sample volume and other global variables
    sample_volume = 0
    total_void_volume = 0
    total_particle_volume = 0  # Changed from fiber to particle for consistency
    total_matrix_volume = 0

    # Open a CSV file to write the thresholds
    with open(csv_file_path, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Image', 'Threshold 1', 'Threshold 2'])  # CSV headers
        
        for idx, (filename, image) in enumerate(images):
            # Convert to grayscale
            gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            
            # Calculate histogram and flatten it
            histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
            
            # Apply K-means clustering to detect thresholds
            thresholds = apply_kmeans_to_histogram(histogram)
            
            # Plot and save the histogram
            plt.figure()
            plt.title(f'Grayscale Histogram for {filename}')
            plt.xlabel('Pixel Values')
            plt.ylabel('Frequency')
            plt.plot(histogram)
            plt.xlim([0, 256])
            plt.axvline(x=thresholds[0], color='red', linestyle='--', label=f'T1: {thresholds[0]:.2f}')
            plt.axvline(x=thresholds[1], color='blue', linestyle='--', label=f'T2: {thresholds[1]:.2f}')
            plt.legend()
            
            # Save the histogram plot
            histogram_filename = f"histogram_{idx + 1}.png"
            plt.savefig(os.path.join(histogram_save_path, histogram_filename))
            plt.close()  # Close the plot to free memory
            
            # Write the filename and thresholds to the CSV file
            writer.writerow([filename, thresholds[0], thresholds[1]])
            print(f"Processed {filename}: Thresholds -> {thresholds}")
            
            # Calculate slice volume : Results in microns
            slice_volume = (gray_image.shape[0] * gray_image.shape[1] * (MICRONS_PER_PIXEL ** 2) * SLICE_THICKNESS_MICRONS)
            sample_volume += slice_volume
            
            # Void detection
            T1 = thresholds[0]
            _, binary_img = cv2.threshold(gray_image, int(T1), 255, cv2.THRESH_BINARY_INV) # Inverse is applied, voids are now white
            num_labels, labels_im = cv2.connectedComponents(binary_img)
            #All connected components and the background are labelled, labelled - 1 = number of voids (one is for the background)
            small_void_areas = 0
            void_areas = []  # List to store large void areas in pixels

            
            for label in range(1, num_labels):  # Ignore label 0 (background)
                region_mask = (labels_im == label)
                region_area = np.sum(region_mask)  # Region area in pixels

                if region_area <= MAX_VOID_AREA_PIXELS:
                    small_void_areas += region_area  # Add small void areas to cumulative sum
                else:
                    void_areas.append(region_area)  # Append large void areas to list
            
            # Convert void areas to microns
            small_void_area_microns = small_void_areas * (MICRONS_PER_PIXEL ** 2) # gets added to total particle area
            void_areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in void_areas]
            
            # Particle detection (changed from fiber to particle for consistency)
            T2 = thresholds[1]
            # Fixed potential precision issue by using T2 directly
            _, binary_particle_img = cv2.threshold(gray_image, int(T2), 255, cv2.THRESH_BINARY)

            # Find contours for particles
            num_labels, labels_particle = cv2.connectedComponents(binary_particle_img)

            particle_area = 0  # Initialize particle area for this image
            for label in range(1, num_labels):  # Iterate over each particle component
                particle_mask = (labels_particle == label)
                particle_area += np.sum(particle_mask)  # Sum the area of all particles
               
            # Void volume calculation : Results in microns
            total_void_area_microns = sum(void_areas_microns) if void_areas_microns else 0
            void_volume_microns = total_void_area_microns * SLICE_THICKNESS_MICRONS
            total_void_volume += void_volume_microns
            
            # Particle volume calculation
            particle_area_microns = particle_area * (MICRONS_PER_PIXEL ** 2) # Convert to microns
            total_particle_area_microns = particle_area_microns + small_void_area_microns
            total_particle_volume_microns = total_particle_area_microns * SLICE_THICKNESS_MICRONS
            total_particle_volume += total_particle_volume_microns

            # Create mask for matrix region (pixel intensity between T1 and T2)
            matrix_mask = (gray_image > T1) & (gray_image <= T2)

            # Calculate the number of pixels in the matrix region
            num_matrix_pixels = np.sum(matrix_mask)

            # Convert pixel count to area and then to volume
            matrix_area = num_matrix_pixels * (MICRONS_PER_PIXEL ** 2)  # Area in microns²
            matrix_volume = matrix_area * SLICE_THICKNESS_MICRONS  # Volume in microns³
            total_matrix_volume += matrix_volume
        # Calculate volume fractions with checks for division by zero
        if sample_volume == 0:
            raise ValueError("Sample volume is zero. Check image dimensions or input data.")
        else:
            void_volume_fraction = (total_void_volume / sample_volume) * 100
            particle_volume_fraction = (total_particle_volume / sample_volume) * 100
            matrix_volume_fraction = (total_matrix_volume / sample_volume) * 100
        
        print(f"Total Sample Volume: {sample_volume:.2f} µm³")
        print(f"Total Void Volume: {total_void_volume:.2f} µm³")
        print(f"Total Particle Volume: {total_particle_volume:.2f} µm³")  # Changed from fiber to particle
        print(f"Void Volume Fraction: {void_volume_fraction:.4f}%")
        print(f"Particle Volume Fraction: {particle_volume_fraction:.4f}%")  # Changed from fiber to particle
        print(f"Matrix Volume Fraction: {matrix_volume_fraction:.4f}%")

# Function to load threshold values from CSV
def load_thresholds_from_csv(csv_file_path):
    """
    Load thresholds from the given CSV file.
    Assumes the CSV file contains 'Image', 'Threshold 1', and 'Threshold 2' columns.
    Finds the largest T1 and smallest T2 for global threshold recommendation.
    """
    try:
        # Load the CSV file into a DataFrame
        thresholds_df = pd.read_csv(csv_file_path, encoding='latin1')

        # Convert to dictionary format { 'image_name': (T1, T2) }
        thresholds_dict = {row['Image']: (row['Threshold 1'], row['Threshold 2']) for _, row in thresholds_df.iterrows()}

        # Extract T1 and T2 values
        t1_values = thresholds_df["Threshold 1"].tolist()
        t2_values = thresholds_df["Threshold 2"].tolist()

        if not t1_values or not t2_values:
            raise ValueError("No valid threshold values found in the CSV.")

        # Find largest T1 and smallest T2
        global_t1 = max(t1_values)  # Largest T1
        global_t2 = min(t2_values)  # Smallest T2

        # Print recommended global threshold
        print("\n*** Recommended Global Threshold ***")
        print(f"T1: {global_t1} (Largest Threshold Value)")
        print(f"T2: {global_t2} (Smallest Threshold Value)\n")

        print("Thresholds loaded successfully from CSV.")
        return thresholds_dict, global_t1, global_t2

    except Exception as e:
        print(f"Error loading CSV file: {e}")
        return None, None, None

# Function to segment voids using threshold T1
def segment_voids_with_t1(image, t1_threshold):
    """Segment voids using the T1 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t1_threshold), 255, cv2.THRESH_BINARY_INV)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to segment particles using threshold T2
def segment_particles_with_t2(image, t2_threshold):
    """Segment particles using the T2 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t2_threshold), 255, cv2.THRESH_BINARY)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to classify voids based on shape and size
def classify_voids(binary_mask, circularity_threshold=0.87, aspect_ratio_threshold=1.5):
    """
    Classify voids into raster voids, spherical voids, and large voids based on shape parameters.
    
    Returns:
    - raster_voids: List of contours for elongated, non-circular voids
    - spherical_voids: List of contours for circular, small voids
    - large_voids: List of contours for voids larger than the size threshold
    """
    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    raster_voids = []
    spherical_voids = []
    large_voids = []

    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Ignore very small voids
            continue

        perimeter = cv2.arcLength(contour, True)
        # Added check for perimeter to avoid division by zero
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensures min(w, h) is at least 1

        # Calculate size in micrometers
        void_diameter = np.sqrt(4 * area / np.pi) * MICRONS_PER_PIXEL  # Convert pixel area to diameter

        # If the void is large (>50µm), classify separately
        if void_diameter > MAX_VOID_SIZE_MICRONS:
            large_voids.append(contour)
        elif circularity > circularity_threshold and aspect_ratio < aspect_ratio_threshold:
            spherical_voids.append(contour)
        else:
            raster_voids.append(contour)

    return raster_voids, spherical_voids, large_voids


def extract_nnd_metrics(void_mask):
    """
    Extract nearest neighbor distance (NND) metrics from a binary void mask.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing NND metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Label connected components
    labels = measure.label(binary_mask)
    regions = measure.regionprops(labels)
    
    # Need at least 2 voids for NND analysis
    if len(regions) < 2:
        return {
            'void_count': len(regions),
            'mean_nnd': None,
            'expected_nnd': None,
            'r_index': None
        }
    
    # Extract centroids
    centroids = np.array([r.centroid for r in regions])
    
    # Build KD-tree and query for nearest neighbors
    tree = cKDTree(centroids)
    distances, _ = tree.query(centroids, k=2)
    nnd = distances[:, 1]  # First column is distance to self (0), second is to nearest neighbor
    
    # Calculate spatial statistics
    area = void_mask.shape[0] * void_mask.shape[1]
    lambda_density = len(centroids) / area
    expected_nnd = 1 / (2 * np.sqrt(lambda_density))
    observed_nnd = np.mean(nnd)
    r_index = observed_nnd / expected_nnd if expected_nnd > 0 else None
    
    return {
        'void_count': len(regions),
        'mean_nnd': observed_nnd,
        'expected_nnd': expected_nnd,
        'r_index': r_index,
        'nnd_values': nnd.tolist(),  # Store all NND values for further analysis
        'nnd_std': np.std(nnd),
        'nnd_min': np.min(nnd),
        'nnd_max': np.max(nnd),
        'nnd_cv': np.std(nnd) / observed_nnd if observed_nnd > 0 else None  # Coefficient of variation
    }

# Add this function after the classify_voids function
def extract_void_morphology_metrics(void_mask):
    """
    Extract comprehensive void morphology, distribution, and orientation metrics.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing void metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Find contours for voids
    contours, _ = cv2.findContours(binary_mask.astype(np.uint8) * 255, 
                                cv2.RETR_EXTERNAL, 
                                cv2.CHAIN_APPROX_SIMPLE)
    
    # Skip if no contours found
    if not contours:
        return None
    
    # Initialize lists for metrics
    circularity_values = []
    aspect_ratio_values = []
    areas = []
    orientations = []
    centers = []
    
    # Process each void
    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Skip very small voids
            continue
            
        perimeter = cv2.arcLength(contour, True)
        # Calculate circularity with safety check for perimeter
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        # Get bounding rectangle for aspect ratio and orientation
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensure denominator is at least 1
        
        # Calculate orientation based on width/height
        orientation_rad = np.arctan2(h, w)
        orientation_deg = orientation_rad * 180 / np.pi
        
        # Get void center
        M = cv2.moments(contour)
        if M["m00"] > 0:
            center_x = int(M["m10"] / M["m00"])
            center_y = int(M["m01"] / M["m00"])
            centers.append((center_x, center_y))
        
        # Store calculated values
        circularity_values.append(circularity)
        aspect_ratio_values.append(aspect_ratio)
        areas.append(area)
        orientations.append(orientation_deg)
    
    # Skip further processing if too few voids
    if len(centers) < 2:
        return {
            'void_count': len(areas),
            'circularity_mean': np.mean(circularity_values) if circularity_values else 0,
            'aspect_ratio_mean': np.mean(aspect_ratio_values) if aspect_ratio_values else 0,
            'area_mean': np.mean(areas) if areas else 0,
        }
    
    # Calculate nearest neighbor distances
    centers_array = np.array(centers)
    tree = cKDTree(centers_array)
    distances, _ = tree.query(centers_array, k=2)  # k=2 because first result is the point itself
    nn_distances = distances[:, 1]  # Take second column (first neighbor)
    
    # Calculate clustering factor (coefficient of variation of nn distances)
    clustering_factor = np.std(nn_distances) / np.mean(nn_distances) if np.mean(nn_distances) > 0 else 0
    
    # Calculate orientation distribution
    orient_0_30_pct = ((np.array(orientations) >= 0) & (np.array(orientations) < 30)).mean() * 100
    orient_30_60_pct = ((np.array(orientations) >= 30) & (np.array(orientations) < 60)).mean() * 100
    orient_60_90_pct = ((np.array(orientations) >= 60) & (np.array(orientations) <= 90)).mean() * 100
    
    # Convert areas to physical units
    areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in areas]
    
    # Calculate orientation tensor (2D simplified version)
    nx_values = np.cos(np.array(orientations) * np.pi / 180)
    ny_values = np.sin(np.array(orientations) * np.pi / 180)
    
    orientation_tensor = np.zeros((2, 2))
    for nx, ny in zip(nx_values, ny_values):
        n = np.array([nx, ny])
        orientation_tensor += np.outer(n, n)
    
    # Normalize tensor
    orientation_tensor /= len(nx_values)
    
    # Get eigenvalues for alignment index
    eigenvalues, _ = np.linalg.eigh(orientation_tensor)
    alignment_index = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0]) if (eigenvalues[1] + eigenvalues[0]) > 0 else 0
    
    # Compile morphology metrics
    morphology_metrics = {
        # Basic void counts and statistics
        'void_count': len(areas),
        
        # Morphology metrics
        'circularity_mean': np.mean(circularity_values),
        'circularity_std': np.std(circularity_values),
        'aspect_ratio_mean': np.mean(aspect_ratio_values),
        'aspect_ratio_std': np.std(aspect_ratio_values),
        'elongated_void_percentage': (np.array(aspect_ratio_values) > 2).mean() * 100,
        'circular_void_percentage': (np.array(circularity_values) > 0.8).mean() * 100,
        
        # Distribution metrics
        'void_area_mean': np.mean(areas_microns),
        'void_area_skewness': stats.skew(areas_microns),
        'void_nn_distance': np.mean(nn_distances),
        'void_clustering': clustering_factor,
        
        # Orientation metrics
        'orientation_0_30_pct': orient_0_30_pct,
        'orientation_30_60_pct': orient_30_60_pct,
        'orientation_60_90_pct': orient_60_90_pct,
        'orientation_alignment': alignment_index
    }
    
    return morphology_metrics

# Function to annotate void segmentation
def annotate_voids(image, binary_mask):
    """
    Annotate the segmented voids on the original image.
    
    Returns:
    - Annotated image
    - Count of each void type
    """
    img_color = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
    raster_voids, spherical_voids, large_voids = classify_voids(binary_mask)

    # Draw raster voids with thinner contours (Green)
    cv2.drawContours(img_color, raster_voids, -1, (0, 255, 0), 1)  # Green for raster voids

    # Draw spherical voids (Red, thinner)
    for contour in spherical_voids:
        cv2.drawContours(img_color, [contour], -1, (0, 0, 255), 1)  # Red contour
        (x, y), radius = cv2.minEnclosingCircle(contour)
        cv2.circle(img_color, (int(x), int(y)), int(radius), (0, 0, 255), 1)  # Red enclosing circle

    # Draw large voids separately (Thicker Red Contours)
    cv2.drawContours(img_color, large_voids, -1, (0, 0, 255), 2)  # Thicker red for large voids

    return img_color, len(raster_voids), len(spherical_voids), len(large_voids)

# Function to create 3D volume from segmented slices
def create_3d_volume(segmented_slices, slice_thickness=SLICE_THICKNESS_MICRONS):
    """
    Create a 3D volume array from a series of 2D segmented slices.
    
    Parameters:
    - segmented_slices: List of binary images where features are marked as 1 or 255
    - slice_thickness: Thickness of each slice in microns
    
    Returns:
    - 3D numpy array representing the stacked volume
    """
    # Check if we have any slices
    if not segmented_slices:
        raise ValueError("No slices provided for 3D volume creation")
    
    # Ensure all slices have the same dimensions
    heights = [slice.shape[0] for slice in segmented_slices]
    widths = [slice.shape[1] for slice in segmented_slices]
    
    if len(set(heights)) > 1 or len(set(widths)) > 1:
        # Resize slices if dimensions vary
        h, w = max(heights), max(widths)
        for i, slice in enumerate(segmented_slices):
            if slice.shape[0] != h or slice.shape[1] != w:
                segmented_slices[i] = cv2.resize(slice, (w, h))
    
    # Normalize slices to binary (0, 1) values
    normalized_slices = []
    for slice in segmented_slices:
        if slice.max() > 1:
            normalized_slices.append(slice > 0)  # Convert to boolean/binary
        else:
            normalized_slices.append(slice)
    
    # Stack slices to create 3D volume
    volume = np.stack(normalized_slices, axis=0)
    
    return volume

def create_mesh_surface(volume, voxel_size):
    """Helper function to create a PyVista mesh from a binary volume"""
    try:
        # Check if volume has any positive values
        if not np.any(volume):
            print("Warning: Volume is empty. Cannot create mesh.")
            return None
        
        # Extract surface using marching cubes
        verts, faces, normals, values = measure.marching_cubes(volume, 0.5)
        
        # Adjust vertices by voxel size - fix dimension ordering
        verts[:, 0] *= voxel_size[0]  # X dimension
        verts[:, 1] *= voxel_size[1]  # Y dimension
        verts[:, 2] *= voxel_size[2]  # Z dimension
        
        # Create PyVista mesh
        faces = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        mesh = pv.PolyData(verts, faces)
        
        # Smooth the mesh for better visualization
        mesh = mesh.smooth(n_iter=15)
        
        return mesh
    except Exception as e:
        print(f"Error creating mesh surface: {str(e)}")
        return None

def create_professional_3d_visualization(void_volume, particle_volume, output_path, 
                                        voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """Create a high-quality 3D visualization of voids, particles, and matrix."""
    try:
        # Suppress warnings
        warnings.filterwarnings("ignore", category=UserWarning)
        warnings.filterwarnings("ignore", category=DeprecationWarning)

        # Check for valid data
        if not np.any(void_volume) and not np.any(particle_volume):
            raise ValueError("Both void and particle volumes are empty.")

        # Configure plotter for compatibility
        pv.set_plot_theme(pv.themes.DocumentTheme())
        plotter = pv.Plotter(
            off_screen=True,
            window_size=[1200, 1000],
            lighting='three lights',
            polygon_smoothing=False
        )

        # Convert voxel size to mm
        voxel_size_mm = (voxel_size[0]/1000, voxel_size[1]/1000, voxel_size[2]/1000)

        # Create and add meshes
        matrix_volume = ~(void_volume | particle_volume)
        
        components = [
            (matrix_volume, 'tan', 'Matrix', 0.6),
            (void_volume, 'blue', 'Voids', 0.4),
            (particle_volume, 'darkred', 'Particles', 0.8)
        ]

        for vol, color, name, opacity in components:
            if np.any(vol):
                mesh = create_mesh_surface(vol, voxel_size)
                if mesh:
                    plotter.add_mesh(mesh, color=color, opacity=opacity,
                                   smooth_shading=False, specular=0.5)

        # Configure axes and labels
        plotter.show_bounds(
            xtitle='X (mm)', 
            ytitle='Y (mm)', 
            ztitle='Z (mm)',
            grid=True,
            location='outer',
            all_edges=True
        )

        # Add legend and title
        plotter.add_legend(
            [('Matrix', 'tan'), ('Voids', 'blue'), ('Particles', 'darkred')],
            size=(0.15, 0.15),
            loc='upper right'
        )
        plotter.add_text("HDPE-HGM Syntactic Foam - 3D Reconstruction", 
                        position='upper_left', font_size=14)

        # Set camera and save output
        plotter.view_isometric()
        plotter.camera.zoom(0.75)
        
        # Create output directory if needed
        output_dir = os.path.dirname(output_path)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)

        # Handle different output formats
        file_ext = os.path.splitext(output_path)[1].lower()
        
        if file_ext == '.html':
            plotter.export_html(output_path)
            html_path = output_path
        else:
            plotter.screenshot(output_path)
            # Create companion HTML file
            html_path = os.path.splitext(output_path)[0] + '.html'
            plotter.export_html(html_path)

        plotter.close()
        print(f"Visualization saved to: {output_path}")
        return output_path, html_path  # Return tuple consistently

    except Exception as e:
        print(f"Visualization error: {str(e)}")
        return None, None  # Return tuple for consistency
        
def generate_mesh_from_volume(volume, output_file, voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """
    Generate a mesh file from a 3D binary volume.
    
    Parameters:
    - volume: 3D binary array
    - output_file: Path to save the mesh file
    - voxel_size: Tuple of (x, y, z) dimensions in microns
    
    Returns:
    - mesh: PyVista mesh object
    """
    # Ensure the volume is binary
    binary_volume = volume > 0
    
    # Check if volume has any positive values before processing
    if not np.any(binary_volume):
        print(f"Warning: Volume for {output_file} is empty. Cannot generate mesh.")
        return None
    
    # Extract surface mesh using marching cubes
    try:
        verts, faces, normals, values = measure.marching_cubes(binary_volume, 0.5)
        
        # Adjust vertices by voxel size - corrected dimension ordering
        verts[:, 0] *= voxel_size[0]
        verts[:, 1] *= voxel_size[1]
        verts[:, 2] *= voxel_size[2]
        
        # Convert faces to the format expected by PyVista
        faces_pv = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        
        # Create PyVista mesh
        mesh = pv.PolyData(verts, faces_pv)
        
        # Make sure the output directory exists
        output_dir = os.path.dirname(output_file)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)
        
        # Save mesh to file
        file_ext = os.path.splitext(output_file)[1].lower()
        
        if file_ext == '.stl':
            mesh.save(output_file, binary=True)
        elif file_ext == '.obj':
            mesh.save(output_file)
        elif file_ext == '.ply':
            mesh.save(output_file)
        else:
            # Default to STL if extension is not recognized
            output_file = os.path.splitext(output_file)[0] + '.stl'
            mesh.save(output_file, binary=True)
        
        print(f"Mesh saved to {output_file}")
        return mesh
        
    except Exception as e:
        print(f"Error generating mesh: {str(e)}")
        return None

# Function to generate separate meshes for voids and particles
def generate_component_meshes(void_volume, particle_volume, output_dir):
    """
    Generate separate mesh files for voids and particles.
    
    Parameters:
    - void_volume: 3D array with voids marked as 1
    - particle_volume: 3D array with particles marked as 1
    - output_dir: Directory to save mesh files
    
    Returns:
    - Dictionary with mesh file paths
    """
    os.makedirs(output_dir, exist_ok=True)
    
    mesh_files = {}
    
    # Generate void mesh
    void_mesh_path = os.path.join(output_dir, "voids_mesh.stl")
    void_mesh = generate_mesh_from_volume(void_volume, void_mesh_path)
    if void_mesh:
        mesh_files['voids'] = void_mesh_path
    
    # Generate particle mesh
    particle_mesh_path = os.path.join(output_dir, "particles_mesh.stl")
    particle_mesh = generate_mesh_from_volume(particle_volume, particle_mesh_path)
    if particle_mesh:
        mesh_files['particles'] = particle_mesh_path
          # Generate combined sample mesh (matrix material)
    matrix_volume = ~(void_volume | particle_volume)
    matrix_mesh_path = os.path.join(output_dir, "matrix_mesh.stl")
    matrix_mesh = generate_mesh_from_volume(matrix_volume, matrix_mesh_path)
    if matrix_mesh:
        mesh_files['matrix'] = matrix_mesh_path
    
    # Generate complete sample mesh (all components together)
    complete_mesh_path = os.path.join(output_dir, "complete_sample.vtm")
    
    # Create multiple meshes in a single VTK file
    blocks = pv.MultiBlock()
    
    if void_mesh:
        blocks.append(void_mesh)
    if particle_mesh:
        blocks.append(particle_mesh)
    if matrix_mesh:
        blocks.append(matrix_mesh)
    
    # Only save if we have at least one valid mesh
    if len(blocks) > 0:
        blocks.save(complete_mesh_path)
        mesh_files['complete'] = complete_mesh_path
    
    return mesh_files

# Function to process a stack of images and create 3D visualizations and meshes
def process_image_stack(images, thresholds_dict, output_dir):
    """
    Process a stack of images to create 3D visualizations and meshes.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Base directory for outputs
    
    Returns:
    - Dictionary with results and file paths
    """
    try:
        # Create output subdirectories
        visualization_dir = os.path.join(output_dir, "3d_visualizations")
        mesh_dir = os.path.join(output_dir, "meshes")
        segmentation_dir = os.path.join(output_dir, "segmented_images")
        morphology_dir = os.path.join(output_dir, "void_morphology")
        
        for directory in [visualization_dir, mesh_dir, segmentation_dir, morphology_dir]:
            os.makedirs(directory, exist_ok=True)
        
        void_slices = []
        particle_slices = []
        segmented_images = []
        
        raster_void_count = 0
        spherical_void_count = 0
        large_void_count = 0
        
        # Lists to store void metrics for each slice
        void_morphology_metrics_per_slice = []
        void_details_list = []
        
        # Process each image
        for idx, (filename, img) in enumerate(images):
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                t1, t2 = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment voids and particles
            void_mask = segment_voids_with_t1(gray_img, t1)
            particle_mask = segment_particles_with_t2(gray_img, t2)
            
            # Annotate voids
            annotated_img, raster_count, spherical_count, large_count = annotate_voids(gray_img, void_mask)
            
            # Extract NND metrics
            nnd_metrics = extract_nnd_metrics(void_mask)
            if nnd_metrics:
                nnd_metrics['filename'] = filename
                nnd_metrics['slice_index'] = idx
                void_morphology_metrics_per_slice.append(nnd_metrics)
            
            # Extract void details for each individual void
            labels = measure.label(void_mask > 0)
            regions = measure.regionprops(labels)
            
            for i, region in enumerate(regions):
                y, x = region.centroid
                width = region.bbox[3] - region.bbox[1]
                height = region.bbox[2] - region.bbox[0]
                area = region.area
                perimeter = region.perimeter if hasattr(region, 'perimeter') else 0
                
                # Calculate circularity and aspect ratio
                circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
                aspect_ratio = max(width, height) / max(min(width, height), 1)
                
                # Calculate orientation
                orientation_rad = np.arctan2(height, width)
                orientation_deg = orientation_rad * 180 / np.pi
                
                # Store void details
                void_details = {
                    'Slice': idx + 1,
                    'Image': filename,
                    'ID': i + 1,
                    'X': x,
                    'Y': y,
                    'Width': width,
                    'Height': height,
                    'Area': area,
                    'Circularity': circularity,
                    'Aspect Ratio': aspect_ratio,
                    'Orientation_deg': orientation_deg
                }
                void_details_list.append(void_details)
            
            # Update counts
            raster_void_count += raster_count
            spherical_void_count += spherical_count
            large_void_count += large_count
            
            # Save segmented image
            segmented_path = os.path.join(segmentation_dir, f"segmented_{filename}")
            cv2.imwrite(segmented_path, annotated_img)
            segmented_images.append(segmented_path)
            
            # Add segmentation masks to slices for 3D reconstruction
            void_slices.append(void_mask)
            particle_slices.append(particle_mask)
        
        # Save void details to CSV
        if void_details_list:
            void_details_df = pd.DataFrame(void_details_list)
            void_details_csv_path = os.path.join(output_dir, "Void_Details.csv")
            void_details_df.to_csv(void_details_csv_path, index=False)
            print(f"Void details saved to: {void_details_csv_path}")
        
        # Check if we have any valid slices
        if not void_slices or not particle_slices:
            raise ValueError("No valid slices found for 3D reconstruction")
        
        # Create 3D volumes
        void_volume = create_3d_volume(void_slices)
        particle_volume = create_3d_volume(particle_slices)
        
        # Create 3D visualization - Fixed to handle tuple return
        viz_path = os.path.join(visualization_dir, "3d_segmentation.png")
        viz_result = create_professional_3d_visualization(
            void_volume, particle_volume, viz_path
        )
        
        if viz_result[0] is None:  # Check if visualization failed
            raise RuntimeError("3D visualization failed")
        
        viz_path, html_path = viz_result
        
        # Generate mesh files
        mesh_files = generate_component_meshes(void_volume, particle_volume, mesh_dir)
        
        # Calculate volume fractions with safe division
        total_volume = void_volume.size
        if total_volume > 0:
            void_volume_fraction = (np.sum(void_volume) / total_volume) * 100
            particle_volume_fraction = (np.sum(particle_volume) / total_volume) * 100
            matrix_volume_fraction = 100 - void_volume_fraction - particle_volume_fraction
        else:
            void_volume_fraction = 0
            particle_volume_fraction = 0
            matrix_volume_fraction = 0
        
        # Aggregate morphology metrics
        aggregate_void_metrics = {}
        
        if void_morphology_metrics_per_slice:
            # Calculate mean values across all slices
            for key in void_morphology_metrics_per_slice[0].keys():
                if (key not in ['filename', 'slice_index', 'nnd_values'] and 
                    isinstance(void_morphology_metrics_per_slice[0][key], (int, float)) and 
                    void_morphology_metrics_per_slice[0][key] is not None):
                    values = [metrics[key] for metrics in void_morphology_metrics_per_slice 
                             if key in metrics and metrics[key] is not None]
                    if values:
                        aggregate_void_metrics[key] = np.mean(values)
        
        # Calculate global morphology statistics from void details
        if void_details_list:
            void_df = pd.DataFrame(void_details_list)
            morphology_stats = {
                'circularity_mean': void_df['Circularity'].mean(),
                'circularity_std': void_df['Circularity'].std(),
                'aspect_ratio_mean': void_df['Aspect Ratio'].mean(),
                'aspect_ratio_std': void_df['Aspect Ratio'].std(),
                'elongated_void_percentage': (void_df['Aspect Ratio'] > 2).mean() * 100,
                'circular_void_percentage': (void_df['Circularity'] > 0.8).mean() * 100,
                'void_area_mean': void_df['Area'].mean(),
                'void_area_skewness': void_df['Area'].skew()
            }
            
            # Calculate orientation distribution
            if 'Orientation_deg' in void_df.columns:
                morphology_stats['orientation_0_30_pct'] = ((void_df['Orientation_deg'] >= 0) & 
                                                          (void_df['Orientation_deg'] < 30)).mean() * 100
                morphology_stats['orientation_30_60_pct'] = ((void_df['Orientation_deg'] >= 30) & 
                                                            (void_df['Orientation_deg'] < 60)).mean() * 100
                morphology_stats['orientation_60_90_pct'] = ((void_df['Orientation_deg'] >= 60) & 
                                                            (void_df['Orientation_deg'] <= 90)).mean() * 100
                
                # Calculate orientation tensor
                rad_angles = void_df['Orientation_deg'] * np.pi / 180
                nx_values = np.cos(rad_angles)
                ny_values = np.sin(rad_angles)
                
                orientation_tensor = np.zeros((2, 2))
                for nx, ny in zip(nx_values, ny_values):
                    n = np.array([nx, ny])
                    orientation_tensor += np.outer(n, n)
                
                if len(nx_values) > 0:
                    orientation_tensor /= len(nx_values)
                    eigenvalues, _ = np.linalg.eigh(orientation_tensor)
                    if (eigenvalues[1] + eigenvalues[0]) > 0:
                        morphology_stats['orientation_alignment'] = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0])
            
            # Update aggregate metrics with morphology stats
            aggregate_void_metrics.update(morphology_stats)
        
        # Compile results
        results = {
            "raster_void_count": raster_void_count,
            "spherical_void_count": spherical_void_count,
            "large_void_count": large_void_count,
            "total_void_count": raster_void_count + spherical_void_count + large_void_count,
            "void_volume_fraction": void_volume_fraction,
            "particle_volume_fraction": particle_volume_fraction,
            "matrix_volume_fraction": matrix_volume_fraction,
            "visualization_path": viz_path,
            "html_visualization_path": html_path,
            "mesh_files": mesh_files,
            "segmented_images": segmented_images,
            "void_volume_shape": void_volume.shape,
            "physical_dimensions_mm": (
                void_volume.shape[2] * MICRONS_PER_PIXEL / 1000,
                void_volume.shape[1] * MICRONS_PER_PIXEL / 1000,
                void_volume.shape[0] * SLICE_THICKNESS_MICRONS / 1000
            ),
            "void_morphology_metrics": aggregate_void_metrics
        }
        
        # Save results to CSV
        results_csv_path = os.path.join(output_dir, "3d_analysis_results.csv")
        with open(results_csv_path, 'w', newline='') as csvfile:
            writer = csv.writer(csvfile)
            writer.writerow(['Parameter', 'Value', 'Units'])
            writer.writerow(['Raster Void Count', raster_void_count, ''])
            writer.writerow(['Spherical Void Count', spherical_void_count, ''])
            writer.writerow(['Large Void Count', large_void_count, ''])
            writer.writerow(['Total Void Count', results["total_void_count"], ''])
            writer.writerow(['Void Volume Fraction', f"{void_volume_fraction:.4f}", '%'])
            writer.writerow(['Particle Volume Fraction', f"{particle_volume_fraction:.4f}", '%'])
            writer.writerow(['Matrix Volume Fraction', f"{matrix_volume_fraction:.4f}", '%'])
            writer.writerow(['X Dimension', f"{results['physical_dimensions_mm'][0]:.4f}", 'mm'])
            writer.writerow(['Y Dimension', f"{results['physical_dimensions_mm'][1]:.4f}", 'mm'])
            writer.writerow(['Z Dimension', f"{results['physical_dimensions_mm'][2]:.4f}", 'mm'])
            
            # Add void morphology metrics to CSV
            writer.writerow(['', '', ''])  # Empty row as separator
            writer.writerow(['Void Morphology and Distribution Metrics', '', ''])
            for key, value in aggregate_void_metrics.items():
                # Format the key for better readability
                formatted_key = ' '.join(word.capitalize() for word in key.split('_'))
                if isinstance(value, (int, float)):
                    writer.writerow([formatted_key, f"{value:.4f}", ''])
        
        # Save detailed morphology metrics to CSV
        if void_morphology_metrics_per_slice:
            morphology_df = pd.DataFrame(void_morphology_metrics_per_slice)
            # Remove the nnd_values column which can be very large
            if 'nnd_values' in morphology_df.columns:
                morphology_df = morphology_df.drop(columns=['nnd_values'])
            morphology_csv_path = os.path.join(morphology_dir, "void_morphology_metrics.csv")
            morphology_df.to_csv(morphology_csv_path, index=False)
            print(f"Void morphology metrics saved to: {morphology_csv_path}")
        
        # Create visualization of void orientation distribution
        if void_details_list:
            void_df = pd.DataFrame(void_details_list)
            if 'Orientation_deg' in void_df.columns:
                # Create orientation rose diagram
                fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})
                
                # Convert to radians and bin the data
                orientations_rad = void_df['Orientation_deg'] * np.pi / 180
                bins = np.linspace(0, np.pi, 19)  # 18 bins covering 0 to 180 degrees
                
                # Create histogram
                hist, bin_edges = np.histogram(orientations_rad, bins=bins)
                
                # Plot rose diagram
                width = np.pi / 18
                bars = ax.bar(bin_edges[:-1], hist, width=width, alpha=0.7)
                
                # Customize appearance
                ax.set_theta_zero_location('N')  # Set 0 degrees at the top
                ax.set_theta_direction(-1)  # Clockwise
                ax.set_xticks(np.linspace(0, np.pi, 7))
                ax.set_xticklabels(['0°', '30°', '60°', '90°', '120°', '150°', '180°'])
                ax.set_title('Void Orientation Distribution', fontsize=16)
                ax.grid(True)
                
                # Save the figure
                orientation_viz_path = os.path.join(morphology_dir, 'void_orientation_distribution.png')
                plt.savefig(orientation_viz_path, dpi=300, bbox_inches='tight')
                plt.close(fig)
                print(f"Void orientation distribution visualization saved to: {orientation_viz_path}")
        
        return results

    except Exception as e:
        print(f"Error in image stack processing: {e}")
        import traceback
        traceback.print_exc()
        return None

# Additional helper functions for specialized analysis
def generate_raster_void_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing raster voids
    - Dictionary of raster void details
    """
    raster_void_slices = []
    raster_void_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            t1, _ = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment voids using T1 threshold
        void_mask = segment_voids_with_t1(gray_img, t1)
        
        # Classify voids to get specifically raster voids
        raster_voids, _, _ = classify_voids(void_mask)
        
        # Create a binary mask for raster voids
        raster_void_mask = np.zeros_like(gray_img)
        cv2.drawContours(raster_void_mask, raster_voids, -1, 255, -1)
        
        # Store slice and details
        raster_void_slices.append(raster_void_mask)
        
        # Collect void details
        void_details = {
            'filename': filename,
            'raster_void_count': len(raster_voids),
            'raster_void_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in raster_voids]
        }
        raster_void_details.append(void_details)
    
    # Create 3D volume
    raster_void_volume = create_3d_volume(raster_void_slices)
    
    return raster_void_volume, raster_void_details

def generate_particle_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing particles
    - Dictionary of particle details
    """
    particle_slices = []
    particle_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            _, t2 = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment particles using T2 threshold
        particle_mask = segment_particles_with_t2(gray_img, t2)
        
        # Find particle contours
        num_labels, labels_particle = cv2.connectedComponents(particle_mask)
        
        # Create list of particle contours
        particle_contours = []
        for label in range(1, num_labels):
            component_mask = (labels_particle == label).astype(np.uint8) * 255
            contours, _ = cv2.findContours(component_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            particle_contours.extend(contours)
        
        # Create binary mask for particles
        particle_binary = np.zeros_like(gray_img)
        cv2.drawContours(particle_binary, particle_contours, -1, 255, -1)
        
        # Store slice and details
        particle_slices.append(particle_binary)
        
        # Collect particle details
        particle_details.append({
            'filename': filename,
            'particle_count': len(particle_contours),
            'particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in particle_contours]
        })
    
    # Create 3D volume
    particle_volume = create_3d_volume(particle_slices)
    
    return particle_volume, particle_details

def generate_raster_void_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "raster_void_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Generate raster void volume
        raster_void_volume, raster_void_details = generate_raster_void_3d_volume(images, thresholds_dict)
        
        # Visualize 3D volume of raster voids
        viz_path = os.path.join(visualization_dir, "raster_voids_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            raster_void_volume, 
            np.zeros_like(raster_void_volume),  # Empty particle volume
            viz_path
        )
        
        return viz_path, raster_void_volume

    except Exception as e:
        print(f"Error generating raster void 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def generate_spherical_particle_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of spherical particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "spherical_particle_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Collect spherical particle slices
        spherical_particle_slices = []
        spherical_particle_details = []

        for filename, img in images:
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                _, t2 = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment particles using T2 threshold
            particle_mask = segment_particles_with_t2(gray_img, t2)
            
            # Classify voids to extract specifically spherical particles
            _, spherical_particles, _ = classify_voids(particle_mask)
            
            # Create a binary mask for spherical particles
            spherical_particle_mask = np.zeros_like(gray_img)
            cv2.drawContours(spherical_particle_mask, spherical_particles, -1, 255, -1)
            
            # Store slice and details
            spherical_particle_slices.append(spherical_particle_mask)
            
            # Collect particle details
            particle_details = {
                'filename': filename,
                'spherical_particle_count': len(spherical_particles),
                'spherical_particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in spherical_particles]
            }
            spherical_particle_details.append(particle_details)
        
        # Create 3D volume
        spherical_particle_volume = create_3d_volume(spherical_particle_slices)
        
        # Visualize 3D volume of spherical particles
        viz_path = os.path.join(visualization_dir, "spherical_particles_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            np.zeros_like(spherical_particle_volume),  # Empty void volume
            spherical_particle_volume, 
            viz_path
        )
        
        return viz_path, spherical_particle_volume

    except Exception as e:
        print(f"Error generating spherical particle 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def visualize_3d_volumes(raster_void_volume, particle_volume, output_dir):
    """
    Visualize and save 3D volumes for raster voids and particles.
    
    Parameters:
    - raster_void_volume: 3D numpy array of raster voids
    - particle_volume: 3D numpy array of particles
    - output_dir: Directory to save visualizations
    
    Returns:
    - Dictionary with visualization paths and details
    """
    # Create output directories
    visualization_dir = os.path.join(output_dir, "3d_component_visualizations")
    mesh_dir = os.path.join(output_dir, "component_meshes")
    
    os.makedirs(visualization_dir, exist_ok=True)
    os.makedirs(mesh_dir, exist_ok=True)
    
    # Visualization for raster voids
    raster_void_viz_path = os.path.join(visualization_dir, "raster_voids_3d.png")
    raster_void_viz_path, _ = create_professional_3d_visualization(
        raster_void_volume, 
        np.zeros_like(raster_void_volume),  # Empty particle volume
        raster_void_viz_path
    )
    
    # Visualization for particles
    particle_viz_path = os.path.join(visualization_dir, "particles_3d.png")
    particle_viz_path, _ = create_professional_3d_visualization(
        np.zeros_like(particle_volume),  # Empty void volume
        particle_volume, 
        particle_viz_path
    )
    
    # Generate mesh files
    raster_void_meshes = generate_component_meshes(
        raster_void_volume, 
        np.zeros_like(raster_void_volume), 
        os.path.join(mesh_dir, "raster_void_meshes")
    )
    
    particle_meshes = generate_component_meshes(
        np.zeros_like(particle_volume),
        particle_volume, 
        os.path.join(mesh_dir, "particle_meshes")
    )
    
    # Calculate volume fractions
    total_raster_void_volume = raster_void_volume.size
    raster_void_volume_fraction = (np.sum(raster_void_volume) / total_raster_void_volume) * 100
    
    total_particle_volume_size = particle_volume.size
    particle_volume_fraction = (np.sum(particle_volume) / total_particle_volume_size) * 100
    
    # Compile results
    results = {
        "raster_void_visualization_path": raster_void_viz_path,
        "particle_visualization_path": particle_viz_path,
        "raster_void_meshes": raster_void_meshes,
        "particle_meshes": particle_meshes,
        "raster_void_volume_fraction": raster_void_volume_fraction,
        "particle_volume_fraction": particle_volume_fraction
    }
    
    return results

if __name__ == "__main__":
    root = tk.Tk()
    app = FoamAnalyzerGUI(root)
    root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox, scrolledtext
from PIL import Image, ImageTk
import os
import pandas as pd
import cv2
import numpy as np
import matplotlib.pyplot as plt
import csv
import warnings
from sklearn.cluster import KMeans
from skimage import measure
from scipy.spatial import cKDTree
from scipy import stats
import pyvista as pv

# REMOVED THE PROBLEMATIC IMPORT - all functions are defined in this file

# Global variable
_root = None

# Constants
MICRONS_PER_PIXEL = 4.89
SLICE_THICKNESS_MICRONS = 4.89
MAX_VOID_SIZE_MICRONS = 50.0
MAX_VOID_SIZE_PIXELS = MAX_VOID_SIZE_MICRONS / MICRONS_PER_PIXEL
MAX_VOID_AREA_PIXELS = np.pi * (MAX_VOID_SIZE_PIXELS / 2) ** 2

class FoamAnalyzerGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("HDPE-HGM Syntactic Foam Analyzer")
        self.root.geometry("1600x1000")
        style = ttk.Style(root)
        style.configure('TFrame', background='#f0f4f8')
        style.configure('TLabel', font=('Times New Roman', 12), background='#f0f4f8')
        style.configure('TButton', font=('Times New Roman', 12))
        style.configure('TEntry', font=('Times New Roman', 12))

        self.annotated_images = []
        self.thresholds_dict = {}

        self.setup_ui()
    
    def setup_ui(self):
        # Top controls frame
        top = ttk.Frame(self.root)
        top.pack(fill='x', pady=5, padx=10)

        # Image folder selection
        ttk.Label(top, text="Image Folder:").grid(row=0, column=0, sticky='w', padx=5)
        self.img_entry = ttk.Entry(top, width=70)
        self.img_entry.grid(row=0, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_image).grid(row=0, column=2, padx=5)

        # CSV file selection
        ttk.Label(top, text="Save Thresholds CSV to:").grid(row=1, column=0, sticky='w', padx=5)
        self.csv_entry = ttk.Entry(top, width=70)
        self.csv_entry.grid(row=1, column=1, sticky='w', padx=5)
        ttk.Button(top, text="Browse", command=self.browse_csv).grid(row=1, column=2, padx=5)

        # Process button
        ttk.Button(top, text="Compute Thresholds", command=self.process_images).grid(row=2, column=1, pady=10)

        # Progress bar
        self.progress = ttk.Progressbar(self.root, mode='indeterminate')
        self.progress.pack(fill='x', pady=5, padx=10)

        # Middle frame for results and controls
        middle_frame = ttk.Frame(self.root)
        middle_frame.pack(fill='both', expand=True, pady=10, padx=10)

        # Left panel for image viewing
        left_panel = ttk.Frame(middle_frame)
        left_panel.pack(side='left', fill='both', expand=True, padx=5)

        # Threshold labels
        threshold_frame = ttk.Frame(left_panel)
        threshold_frame.pack(fill='x', pady=5)
        
        self.t1label = ttk.Label(threshold_frame, text="Global T1: Not calculated")
        self.t1label.pack(side='left', padx=10)
        
        self.t2label = ttk.Label(threshold_frame, text="Global T2: Not calculated")
        self.t2label.pack(side='left', padx=10)

        # Image display
        image_frame = ttk.Frame(left_panel)
        image_frame.pack(fill='both', expand=True, pady=5)
        
        self.ct_img_label = ttk.Label(image_frame, text="No images loaded", 
                                     relief='sunken', borderwidth=2)
        self.ct_img_label.pack(fill='both', expand=True)

        # Image slider
        slider_frame = ttk.Frame(left_panel)
        slider_frame.pack(fill='x', pady=5)
        
        ttk.Label(slider_frame, text="Image:").pack(side='left')
        self.slider = ttk.Scale(slider_frame, from_=0, to=0, orient='horizontal',
                               command=self.update_image)
        self.slider.pack(side='left', fill='x', expand=True, padx=5)

        # Right panel for 3D visualization and controls
        right_panel = ttk.Frame(middle_frame)
        right_panel.pack(side='right', fill='both', expand=True, padx=5)

        # 3D visualization buttons
        button_frame = ttk.Frame(right_panel)
        button_frame.pack(fill='x', pady=5)
        
        ttk.Button(button_frame, text="Show Voids 3D", 
                  command=self.show_voids3d).pack(side='left', padx=5)
        ttk.Button(button_frame, text="Show Particles 3D", 
                  command=self.show_particles3d).pack(side='left', padx=5)

        # 3D display area
        self.static_3d_label = ttk.Label(right_panel, text="3D Visualization Area", 
                                        relief='sunken', borderwidth=2)
        self.static_3d_label.pack(fill='both', expand=True, pady=5)

        # Volume fraction display
        self.vflabel = ttk.Label(right_panel, 
                                text="VF% - Process images to see results",
                                font=('Times New Roman', 10))
        self.vflabel.pack(fill='x', pady=5)
        
    def browse_image(self):
        """Browse for image folder."""
        path = filedialog.askdirectory(title="Select Image Folder")
        if path:
            self.img_entry.delete(0, 'end')
            self.img_entry.insert(0, path)

    def browse_csv(self):
        """Browse for CSV file location to save computed thresholds."""
        path = filedialog.asksaveasfilename(
            title="Where to Save Computed Thresholds CSV",
            defaultextension=".csv",
            filetypes=[("CSV files", "*.csv")]
        )
        if path:
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, path)

    def process_images(self):
        """Process images and compute thresholds."""
        folder = self.img_entry.get()
        csv_file = self.csv_entry.get()
        
        if not folder:
            messagebox.showerror("Error", "Please select an image folder")
            return

        # Always prompt for CSV save location if not provided
        if not csv_file:
            csv_file = filedialog.asksaveasfilename(
                defaultextension=".csv",
                filetypes=[("CSV files", "*.csv")],
                title="Save computed thresholds CSV as..."
            )
            if not csv_file:
                messagebox.showerror("Error", "Please specify where to save the thresholds CSV")
                return
            self.csv_entry.delete(0, 'end')
            self.csv_entry.insert(0, csv_file)

        # Start progress bar
        self.progress.start()
        self.root.update()

        try:
            # Step 1: Load images from folder
            images = load_images_from_folder(folder)
            if not images:
                raise ValueError("No valid images found in the selected folder")
            
            messagebox.showinfo("Info", f"Found {len(images)} images. Now computing thresholds...")
            self.root.update()
            
            # Step 2: COMPUTE thresholds and save to CSV
            process_images_and_save_results(images, "histograms", csv_file)
            
            messagebox.showinfo("Success", f"Thresholds computed and saved to:\n{csv_file}")
            
            # Step 3: Load the computed thresholds back
            threshold_result = load_thresholds_from_csv(csv_file)
            if threshold_result[0] is None:
                raise ValueError("Failed to load computed thresholds from CSV")
            
            self.thresholds_dict, g1, g2 = threshold_result
            self.t1label.config(text=f"Global T1: {g1:.2f}")
            self.t2label.config(text=f"Global T2: {g2:.2f}")

            # Step 4: Load images for display
            self.annotated_images = []
            for filename, img in images:
                # Convert to display format
                if len(img.shape) == 3:
                    display_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                else:
                    display_img = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
                
                pil_img = Image.fromarray(display_img).resize((400, 400))
                self.annotated_images.append(pil_img)
            
            # Update slider and display first image
            if self.annotated_images:
                self.slider.config(to=len(self.annotated_images)-1)
                self.update_image(0)
            
            messagebox.showinfo("Complete", 
                               f"Processing completed successfully!\n\n"
                               f"• {len(images)} images processed\n"
                               f"• Thresholds computed and saved\n"
                               f"• Histogram plots saved to 'histograms' folder\n"
                               f"• Ready for 3D analysis!")
            
        except Exception as e:
            messagebox.showerror("Error", f"Processing failed: {str(e)}")
        finally:
            self.progress.stop()

    def update_image(self, val):
        """Update displayed image based on slider value."""
        try:
            idx = int(float(val))
            if 0 <= idx < len(self.annotated_images):
                img = self.annotated_images[idx]
                imgtk = ImageTk.PhotoImage(img)
                self.ct_img_label.config(image=imgtk, text="")
                self.ct_img_label.image = imgtk  # Keep a reference
        except (ValueError, IndexError):
            pass  # Ignore invalid values

    def show_voids3d(self):
        """Generate and display 3D void visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="")
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")

    def show_particles3d(self):
        """Generate and display 3D particle visualization."""
        if not hasattr(self, 'thresholds_dict') or not self.thresholds_dict:
            messagebox.showerror("Error", "Please process images first")
            return
        
        folder = self.img_entry.get()
        if not folder:
            messagebox.showerror("Error", "No image folder selected")
            return
        
        try:
            # Load images and process
            images = load_images_from_folder(folder)
            results = process_image_stack(
                images,
                self.thresholds_dict,
                "output"
            )
            
            if results and results.get('visualization_path'):
                # Display the visualization
                img = Image.open(results['visualization_path']).resize((600, 600))
                imgtk = ImageTk.PhotoImage(img)
                self.static_3d_label.config(image=imgtk, text="Particles 3D", compound='top')
                self.static_3d_label.image = imgtk
                
                # Update metrics display
                self.vflabel.config(
                    text=f"VF% - Void: {results['void_volume_fraction']:.2f}% | "
                         f"Particle: {results['particle_volume_fraction']:.2f}% | "
                         f"Matrix: {results['matrix_volume_fraction']:.2f}%"
                )
            else:
                messagebox.showerror("Error", "Failed to generate 3D visualization")
                
        except Exception as e:
            messagebox.showerror("Error", f"3D visualization failed: {str(e)}")


# MOVED get_root() OUTSIDE THE CLASS - FIXED INDENTATION
def get_root():
    global _root
    if _root is None:
        _root = tk.Tk()
        _root.withdraw()
    return _root

# REMOVED unused select_folder and select_csv_file functions to avoid confusion

def load_images_from_folder(folder_path):
    """Load images from the specified folder."""
    images = []
    for filename in sorted(os.listdir(folder_path)):
        if filename.lower().endswith(('.tif', '.tiff', '.png', '.jpg', '.jpeg')):
            img_path = os.path.join(folder_path, filename)
            img = cv2.imread(img_path)
            if img is not None:
                images.append((filename, img))  # Append as (filename, image)
    return images

def calculate_histogram(gray_image):
    """Calculate and return the histogram of a grayscale image."""
    histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
    return histogram

def calculate_epsilon(histogram, sample_size=10000):
    """Compute epsilon for neighborhood initialization (approximate mean pairwise distance)."""
    pixel_values = np.arange(256)
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot calculate epsilon.")
    
    probabilities = histogram / histogram_sum  # Normalize histogram to get probabilities
    sample_size = int(min(sample_size, histogram_sum))

    sampled_data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    mean_distance = np.mean(np.abs(np.diff(np.sort(sampled_data))))
    return mean_distance

def initialize_kmeans_centers(histogram, k, epsilon=None):
    pixel_values = np.arange(len(histogram))
    histogram_sum = histogram.sum()

    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot initialize K-means centers.")

    probabilities = histogram / histogram_sum  # Normalize histogram
    sample_size = int(min(10000, histogram_sum))
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)

    if epsilon is None:
        epsilon = calculate_epsilon(histogram)

    centers = []
    for _ in range(k):
        if len(centers) == 0:
            first_center = np.argmax(histogram)
            centers.append(first_center)
        else:
            distances = np.min([np.abs(data - center) for center in centers], axis=0)
            if np.max(distances) <= epsilon:
                new_center = data[np.argmax(distances)]
            else:
                new_center = data[np.argmax(distances)]
            if new_center not in centers:
                centers.append(new_center)
            else:
                distances[np.argmax(distances)] = 0
                next_center = data[np.argmax(distances)]
                centers.append(next_center)
    return np.array(sorted(centers))

def apply_kmeans_to_histogram(histogram, k=3):
    """Apply K-means clustering using custom initialization."""
    histogram_sum = histogram.sum()
    
    if histogram_sum == 0:
        raise ValueError("Histogram sum is zero. Cannot proceed with K-means clustering.")
    
    pixel_values = np.arange(len(histogram))
    probabilities = histogram / histogram_sum
    sample_size = min(10000, int(histogram_sum))
    
    data = np.random.choice(pixel_values, size=sample_size, p=probabilities)
    initial_centers = initialize_kmeans_centers(histogram, k)
    
    kmeans = KMeans(n_clusters=k, init=initial_centers.reshape(-1, 1), n_init=1, random_state=42)
    kmeans.fit(data.reshape(-1, 1))

    centers = np.sort(kmeans.cluster_centers_.flatten())
    thresholds = [(centers[i] + centers[i + 1]) / 2 for i in range(len(centers) - 1)]
    return thresholds

def process_images_and_save_results(images, histogram_save_path, csv_file_path):
    os.makedirs(histogram_save_path, exist_ok=True)  # Ensure the directory exists
    
    # Initialize sample volume and other global variables
    sample_volume = 0
    total_void_volume = 0
    total_particle_volume = 0  # Changed from fiber to particle for consistency
    total_matrix_volume = 0

    # Open a CSV file to write the thresholds
    with open(csv_file_path, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Image', 'Threshold 1', 'Threshold 2'])  # CSV headers
        
        for idx, (filename, image) in enumerate(images):
            # Convert to grayscale
            gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            
            # Calculate histogram and flatten it
            histogram = cv2.calcHist([gray_image], [0], None, [256], [0, 256]).flatten()
            
            # Apply K-means clustering to detect thresholds
            thresholds = apply_kmeans_to_histogram(histogram)
            
            # Plot and save the histogram
            plt.figure()
            plt.title(f'Grayscale Histogram for {filename}')
            plt.xlabel('Pixel Values')
            plt.ylabel('Frequency')
            plt.plot(histogram)
            plt.xlim([0, 256])
            plt.axvline(x=thresholds[0], color='red', linestyle='--', label=f'T1: {thresholds[0]:.2f}')
            plt.axvline(x=thresholds[1], color='blue', linestyle='--', label=f'T2: {thresholds[1]:.2f}')
            plt.legend()
            
            # Save the histogram plot
            histogram_filename = f"histogram_{idx + 1}.png"
            plt.savefig(os.path.join(histogram_save_path, histogram_filename))
            plt.close()  # Close the plot to free memory
            
            # Write the filename and thresholds to the CSV file
            writer.writerow([filename, thresholds[0], thresholds[1]])
            print(f"Processed {filename}: Thresholds -> {thresholds}")
            
            # Calculate slice volume : Results in microns
            slice_volume = (gray_image.shape[0] * gray_image.shape[1] * (MICRONS_PER_PIXEL ** 2) * SLICE_THICKNESS_MICRONS)
            sample_volume += slice_volume
            
            # Void detection
            T1 = thresholds[0]
            _, binary_img = cv2.threshold(gray_image, int(T1), 255, cv2.THRESH_BINARY_INV) # Inverse is applied, voids are now white
            num_labels, labels_im = cv2.connectedComponents(binary_img)
            #All connected components and the background are labelled, labelled - 1 = number of voids (one is for the background)
            small_void_areas = 0
            void_areas = []  # List to store large void areas in pixels

            
            for label in range(1, num_labels):  # Ignore label 0 (background)
                region_mask = (labels_im == label)
                region_area = np.sum(region_mask)  # Region area in pixels

                if region_area <= MAX_VOID_AREA_PIXELS:
                    small_void_areas += region_area  # Add small void areas to cumulative sum
                else:
                    void_areas.append(region_area)  # Append large void areas to list
            
            # Convert void areas to microns
            small_void_area_microns = small_void_areas * (MICRONS_PER_PIXEL ** 2) # gets added to total particle area
            void_areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in void_areas]
            
            # Particle detection (changed from fiber to particle for consistency)
            T2 = thresholds[1]
            # Fixed potential precision issue by using T2 directly
            _, binary_particle_img = cv2.threshold(gray_image, int(T2), 255, cv2.THRESH_BINARY)

            # Find contours for particles
            num_labels, labels_particle = cv2.connectedComponents(binary_particle_img)

            particle_area = 0  # Initialize particle area for this image
            for label in range(1, num_labels):  # Iterate over each particle component
                particle_mask = (labels_particle == label)
                particle_area += np.sum(particle_mask)  # Sum the area of all particles
               
            # Void volume calculation : Results in microns
            total_void_area_microns = sum(void_areas_microns) if void_areas_microns else 0
            void_volume_microns = total_void_area_microns * SLICE_THICKNESS_MICRONS
            total_void_volume += void_volume_microns
            
            # Particle volume calculation
            particle_area_microns = particle_area * (MICRONS_PER_PIXEL ** 2) # Convert to microns
            total_particle_area_microns = particle_area_microns + small_void_area_microns
            total_particle_volume_microns = total_particle_area_microns * SLICE_THICKNESS_MICRONS
            total_particle_volume += total_particle_volume_microns

            # Create mask for matrix region (pixel intensity between T1 and T2)
            matrix_mask = (gray_image > T1) & (gray_image <= T2)

            # Calculate the number of pixels in the matrix region
            num_matrix_pixels = np.sum(matrix_mask)

            # Convert pixel count to area and then to volume
            matrix_area = num_matrix_pixels * (MICRONS_PER_PIXEL ** 2)  # Area in microns²
            matrix_volume = matrix_area * SLICE_THICKNESS_MICRONS  # Volume in microns³
            total_matrix_volume += matrix_volume
        # Calculate volume fractions with checks for division by zero
        if sample_volume == 0:
            raise ValueError("Sample volume is zero. Check image dimensions or input data.")
        else:
            void_volume_fraction = (total_void_volume / sample_volume) * 100
            particle_volume_fraction = (total_particle_volume / sample_volume) * 100
            matrix_volume_fraction = (total_matrix_volume / sample_volume) * 100
        
        print(f"Total Sample Volume: {sample_volume:.2f} µm³")
        print(f"Total Void Volume: {total_void_volume:.2f} µm³")
        print(f"Total Particle Volume: {total_particle_volume:.2f} µm³")  # Changed from fiber to particle
        print(f"Void Volume Fraction: {void_volume_fraction:.4f}%")
        print(f"Particle Volume Fraction: {particle_volume_fraction:.4f}%")  # Changed from fiber to particle
        print(f"Matrix Volume Fraction: {matrix_volume_fraction:.4f}%")

# Function to load threshold values from CSV
def load_thresholds_from_csv(csv_file_path):
    """
    Load thresholds from the given CSV file.
    Assumes the CSV file contains 'Image', 'Threshold 1', and 'Threshold 2' columns.
    Finds the largest T1 and smallest T2 for global threshold recommendation.
    """
    try:
        # Load the CSV file into a DataFrame
        thresholds_df = pd.read_csv(csv_file_path, encoding='latin1')

        # Convert to dictionary format { 'image_name': (T1, T2) }
        thresholds_dict = {row['Image']: (row['Threshold 1'], row['Threshold 2']) for _, row in thresholds_df.iterrows()}

        # Extract T1 and T2 values
        t1_values = thresholds_df["Threshold 1"].tolist()
        t2_values = thresholds_df["Threshold 2"].tolist()

        if not t1_values or not t2_values:
            raise ValueError("No valid threshold values found in the CSV.")

        # Find largest T1 and smallest T2
        global_t1 = max(t1_values)  # Largest T1
        global_t2 = min(t2_values)  # Smallest T2

        # Print recommended global threshold
        print("\n*** Recommended Global Threshold ***")
        print(f"T1: {global_t1} (Largest Threshold Value)")
        print(f"T2: {global_t2} (Smallest Threshold Value)\n")

        print("Thresholds loaded successfully from CSV.")
        return thresholds_dict, global_t1, global_t2

    except Exception as e:
        print(f"Error loading CSV file: {e}")
        return None, None, None

# Function to segment voids using threshold T1
def segment_voids_with_t1(image, t1_threshold):
    """Segment voids using the T1 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t1_threshold), 255, cv2.THRESH_BINARY_INV)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to segment particles using threshold T2
def segment_particles_with_t2(image, t2_threshold):
    """Segment particles using the T2 threshold value."""
    _, binary_mask = cv2.threshold(image, int(t2_threshold), 255, cv2.THRESH_BINARY)
    return cv2.morphologyEx(binary_mask, cv2.MORPH_OPEN, np.ones((3,3), np.uint8))

# Function to classify voids based on shape and size
def classify_voids(binary_mask, circularity_threshold=0.87, aspect_ratio_threshold=1.5):
    """
    Classify voids into raster voids, spherical voids, and large voids based on shape parameters.
    
    Returns:
    - raster_voids: List of contours for elongated, non-circular voids
    - spherical_voids: List of contours for circular, small voids
    - large_voids: List of contours for voids larger than the size threshold
    """
    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    raster_voids = []
    spherical_voids = []
    large_voids = []

    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Ignore very small voids
            continue

        perimeter = cv2.arcLength(contour, True)
        # Added check for perimeter to avoid division by zero
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensures min(w, h) is at least 1

        # Calculate size in micrometers
        void_diameter = np.sqrt(4 * area / np.pi) * MICRONS_PER_PIXEL  # Convert pixel area to diameter

        # If the void is large (>50µm), classify separately
        if void_diameter > MAX_VOID_SIZE_MICRONS:
            large_voids.append(contour)
        elif circularity > circularity_threshold and aspect_ratio < aspect_ratio_threshold:
            spherical_voids.append(contour)
        else:
            raster_voids.append(contour)

    return raster_voids, spherical_voids, large_voids

def extract_nnd_metrics(void_mask):
    """
    Extract nearest neighbor distance (NND) metrics from a binary void mask.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing NND metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Label connected components
    labels = measure.label(binary_mask)
    regions = measure.regionprops(labels)
    
    # Need at least 2 voids for NND analysis
    if len(regions) < 2:
        return {
            'void_count': len(regions),
            'mean_nnd': None,
            'expected_nnd': None,
            'r_index': None
        }
    
    # Extract centroids
    centroids = np.array([r.centroid for r in regions])
    
    # Build KD-tree and query for nearest neighbors
    tree = cKDTree(centroids)
    distances, _ = tree.query(centroids, k=2)
    nnd = distances[:, 1]  # First column is distance to self (0), second is to nearest neighbor
    
    # Calculate spatial statistics
    area = void_mask.shape[0] * void_mask.shape[1]
    lambda_density = len(centroids) / area
    expected_nnd = 1 / (2 * np.sqrt(lambda_density))
    observed_nnd = np.mean(nnd)
    r_index = observed_nnd / expected_nnd if expected_nnd > 0 else None
    
    return {
        'void_count': len(regions),
        'mean_nnd': observed_nnd,
        'expected_nnd': expected_nnd,
        'r_index': r_index,
        'nnd_values': nnd.tolist(),  # Store all NND values for further analysis
        'nnd_std': np.std(nnd),
        'nnd_min': np.min(nnd),
        'nnd_max': np.max(nnd),
        'nnd_cv': np.std(nnd) / observed_nnd if observed_nnd > 0 else None  # Coefficient of variation
    }

# Add this function after the classify_voids function
def extract_void_morphology_metrics(void_mask):
    """
    Extract comprehensive void morphology, distribution, and orientation metrics.
    
    Parameters:
    - void_mask: Binary mask with voids as 1 or 255
    
    Returns:
    - Dictionary containing void metrics
    """
    # Ensure binary mask
    binary_mask = void_mask > 0
    
    # Find contours for voids
    contours, _ = cv2.findContours(binary_mask.astype(np.uint8) * 255, 
                                cv2.RETR_EXTERNAL, 
                                cv2.CHAIN_APPROX_SIMPLE)
    
    # Skip if no contours found
    if not contours:
        return None
    
    # Initialize lists for metrics
    circularity_values = []
    aspect_ratio_values = []
    areas = []
    orientations = []
    centers = []
    
    # Process each void
    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 5:  # Skip very small voids
            continue
            
        perimeter = cv2.arcLength(contour, True)
        # Calculate circularity with safety check for perimeter
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        
        # Get bounding rectangle for aspect ratio and orientation
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = max(w, h) / max(min(w, h), 1)  # Ensure denominator is at least 1
        
        # Calculate orientation based on width/height
        orientation_rad = np.arctan2(h, w)
        orientation_deg = orientation_rad * 180 / np.pi
        
        # Get void center
        M = cv2.moments(contour)
        if M["m00"] > 0:
            center_x = int(M["m10"] / M["m00"])
            center_y = int(M["m01"] / M["m00"])
            centers.append((center_x, center_y))
        
        # Store calculated values
        circularity_values.append(circularity)
        aspect_ratio_values.append(aspect_ratio)
        areas.append(area)
        orientations.append(orientation_deg)
    
    # Skip further processing if too few voids
    if len(centers) < 2:
        return {
            'void_count': len(areas),
            'circularity_mean': np.mean(circularity_values) if circularity_values else 0,
            'aspect_ratio_mean': np.mean(aspect_ratio_values) if aspect_ratio_values else 0,
            'area_mean': np.mean(areas) if areas else 0,
        }
    
    # Calculate nearest neighbor distances
    centers_array = np.array(centers)
    tree = cKDTree(centers_array)
    distances, _ = tree.query(centers_array, k=2)  # k=2 because first result is the point itself
    nn_distances = distances[:, 1]  # Take second column (first neighbor)
    
    # Calculate clustering factor (coefficient of variation of nn distances)
    clustering_factor = np.std(nn_distances) / np.mean(nn_distances) if np.mean(nn_distances) > 0 else 0
    
    # Calculate orientation distribution
    orient_0_30_pct = ((np.array(orientations) >= 0) & (np.array(orientations) < 30)).mean() * 100
    orient_30_60_pct = ((np.array(orientations) >= 30) & (np.array(orientations) < 60)).mean() * 100
    orient_60_90_pct = ((np.array(orientations) >= 60) & (np.array(orientations) <= 90)).mean() * 100
    
    # Convert areas to physical units
    areas_microns = [area * (MICRONS_PER_PIXEL ** 2) for area in areas]
    
    # Calculate orientation tensor (2D simplified version)
    nx_values = np.cos(np.array(orientations) * np.pi / 180)
    ny_values = np.sin(np.array(orientations) * np.pi / 180)
    
    orientation_tensor = np.zeros((2, 2))
    for nx, ny in zip(nx_values, ny_values):
        n = np.array([nx, ny])
        orientation_tensor += np.outer(n, n)
    
    # Normalize tensor
    orientation_tensor /= len(nx_values)
    
    # Get eigenvalues for alignment index
    eigenvalues, _ = np.linalg.eigh(orientation_tensor)
    alignment_index = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0]) if (eigenvalues[1] + eigenvalues[0]) > 0 else 0
    
    # Compile morphology metrics
    morphology_metrics = {
        # Basic void counts and statistics
        'void_count': len(areas),
        
        # Morphology metrics
        'circularity_mean': np.mean(circularity_values),
        'circularity_std': np.std(circularity_values),
        'aspect_ratio_mean': np.mean(aspect_ratio_values),
        'aspect_ratio_std': np.std(aspect_ratio_values),
        'elongated_void_percentage': (np.array(aspect_ratio_values) > 2).mean() * 100,
        'circular_void_percentage': (np.array(circularity_values) > 0.8).mean() * 100,
        
        # Distribution metrics
        'void_area_mean': np.mean(areas_microns),
        'void_area_skewness': stats.skew(areas_microns),
        'void_nn_distance': np.mean(nn_distances),
        'void_clustering': clustering_factor,
        
        # Orientation metrics
        'orientation_0_30_pct': orient_0_30_pct,
        'orientation_30_60_pct': orient_30_60_pct,
        'orientation_60_90_pct': orient_60_90_pct,
        'orientation_alignment': alignment_index
    }
    
    return morphology_metrics

# Function to annotate void segmentation
def annotate_voids(image, binary_mask):
    """
    Annotate the segmented voids on the original image.
    
    Returns:
    - Annotated image
    - Count of each void type
    """
    img_color = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
    raster_voids, spherical_voids, large_voids = classify_voids(binary_mask)

    # Draw raster voids with thinner contours (Green)
    cv2.drawContours(img_color, raster_voids, -1, (0, 255, 0), 1)  # Green for raster voids

    # Draw spherical voids (Red, thinner)
    for contour in spherical_voids:
        cv2.drawContours(img_color, [contour], -1, (0, 0, 255), 1)  # Red contour
        (x, y), radius = cv2.minEnclosingCircle(contour)
        cv2.circle(img_color, (int(x), int(y)), int(radius), (0, 0, 255), 1)  # Red enclosing circle

    # Draw large voids separately (Thicker Red Contours)
    cv2.drawContours(img_color, large_voids, -1, (0, 0, 255), 2)  # Thicker red for large voids

    return img_color, len(raster_voids), len(spherical_voids), len(large_voids)

# Function to create 3D volume from segmented slices
def create_3d_volume(segmented_slices, slice_thickness=SLICE_THICKNESS_MICRONS):
    """
    Create a 3D volume array from a series of 2D segmented slices.
    
    Parameters:
    - segmented_slices: List of binary images where features are marked as 1 or 255
    - slice_thickness: Thickness of each slice in microns
    
    Returns:
    - 3D numpy array representing the stacked volume
    """
    # Check if we have any slices
    if not segmented_slices:
        raise ValueError("No slices provided for 3D volume creation")
    
    # Ensure all slices have the same dimensions
    heights = [slice.shape[0] for slice in segmented_slices]
    widths = [slice.shape[1] for slice in segmented_slices]
    
    if len(set(heights)) > 1 or len(set(widths)) > 1:
        # Resize slices if dimensions vary
        h, w = max(heights), max(widths)
        for i, slice in enumerate(segmented_slices):
            if slice.shape[0] != h or slice.shape[1] != w:
                segmented_slices[i] = cv2.resize(slice, (w, h))
    
    # Normalize slices to binary (0, 1) values
    normalized_slices = []
    for slice in segmented_slices:
        if slice.max() > 1:
            normalized_slices.append(slice > 0)  # Convert to boolean/binary
        else:
            normalized_slices.append(slice)
    
    # Stack slices to create 3D volume
    volume = np.stack(normalized_slices, axis=0)
    
    return volume

def create_mesh_surface(volume, voxel_size):
    """Helper function to create a PyVista mesh from a binary volume"""
    try:
        # Check if volume has any positive values
        if not np.any(volume):
            print("Warning: Volume is empty. Cannot create mesh.")
            return None
        
        # Extract surface using marching cubes
        verts, faces, normals, values = measure.marching_cubes(volume, 0.5)
        
        # Adjust vertices by voxel size - fix dimension ordering
        verts[:, 0] *= voxel_size[0]  # X dimension
        verts[:, 1] *= voxel_size[1]  # Y dimension
        verts[:, 2] *= voxel_size[2]  # Z dimension
        
        # Create PyVista mesh
        faces = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        mesh = pv.PolyData(verts, faces)
        
        # Smooth the mesh for better visualization
        mesh = mesh.smooth(n_iter=15)
        
        return mesh
    except Exception as e:
        print(f"Error creating mesh surface: {str(e)}")
        return None

def create_professional_3d_visualization(void_volume, particle_volume, output_path, 
                                        voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """Create a high-quality 3D visualization of voids, particles, and matrix."""
    try:
        # Suppress warnings
        warnings.filterwarnings("ignore", category=UserWarning)
        warnings.filterwarnings("ignore", category=DeprecationWarning)

        # Check for valid data
        if not np.any(void_volume) and not np.any(particle_volume):
            raise ValueError("Both void and particle volumes are empty.")

        # Configure plotter for compatibility
        pv.set_plot_theme(pv.themes.DocumentTheme())
        plotter = pv.Plotter(
            off_screen=True,
            window_size=[1200, 1000],
            lighting='three lights',
            polygon_smoothing=False
        )

        # Convert voxel size to mm
        voxel_size_mm = (voxel_size[0]/1000, voxel_size[1]/1000, voxel_size[2]/1000)

        # Create and add meshes
        matrix_volume = ~(void_volume | particle_volume)
        
        components = [
            (matrix_volume, 'tan', 'Matrix', 0.6),
            (void_volume, 'blue', 'Voids', 0.4),
            (particle_volume, 'darkred', 'Particles', 0.8)
        ]

        for vol, color, name, opacity in components:
            if np.any(vol):
                mesh = create_mesh_surface(vol, voxel_size)
                if mesh:
                    plotter.add_mesh(mesh, color=color, opacity=opacity,
                                   smooth_shading=False, specular=0.5)

        # Configure axes and labels
        plotter.show_bounds(
            xtitle='X (mm)', 
            ytitle='Y (mm)', 
            ztitle='Z (mm)',
            grid=True,
            location='outer',
            all_edges=True
        )

        # Add legend and title
        plotter.add_legend(
            [('Matrix', 'tan'), ('Voids', 'blue'), ('Particles', 'darkred')],
            size=(0.15, 0.15),
            loc='upper right'
        )
        plotter.add_text("HDPE-HGM Syntactic Foam - 3D Reconstruction", 
                        position='upper_left', font_size=14)

        # Set camera and save output
        plotter.view_isometric()
        plotter.camera.zoom(0.75)
        
        # Create output directory if needed
        output_dir = os.path.dirname(output_path)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)

        # Handle different output formats
        file_ext = os.path.splitext(output_path)[1].lower()
        
        if file_ext == '.html':
            plotter.export_html(output_path)
            html_path = output_path
        else:
            plotter.screenshot(output_path)
            # Create companion HTML file
            html_path = os.path.splitext(output_path)[0] + '.html'
            plotter.export_html(html_path)

        plotter.close()
        print(f"Visualization saved to: {output_path}")
        return output_path, html_path  # Return tuple consistently

    except Exception as e:
        print(f"Visualization error: {str(e)}")
        return None, None  # Return tuple for consistency
        
def generate_mesh_from_volume(volume, output_file, voxel_size=(MICRONS_PER_PIXEL, MICRONS_PER_PIXEL, SLICE_THICKNESS_MICRONS)):
    """
    Generate a mesh file from a 3D binary volume.
    
    Parameters:
    - volume: 3D binary array
    - output_file: Path to save the mesh file
    - voxel_size: Tuple of (x, y, z) dimensions in microns
    
    Returns:
    - mesh: PyVista mesh object
    """
    # Ensure the volume is binary
    binary_volume = volume > 0
    
    # Check if volume has any positive values before processing
    if not np.any(binary_volume):
        print(f"Warning: Volume for {output_file} is empty. Cannot generate mesh.")
        return None
    
    # Extract surface mesh using marching cubes
    try:
        verts, faces, normals, values = measure.marching_cubes(binary_volume, 0.5)
        
        # Adjust vertices by voxel size - corrected dimension ordering
        verts[:, 0] *= voxel_size[0]
        verts[:, 1] *= voxel_size[1]
        verts[:, 2] *= voxel_size[2]
        
        # Convert faces to the format expected by PyVista
        faces_pv = np.column_stack((np.full(len(faces), 3), faces)).flatten()
        
        # Create PyVista mesh
        mesh = pv.PolyData(verts, faces_pv)
        
        # Make sure the output directory exists
        output_dir = os.path.dirname(output_file)
        if output_dir and not os.path.exists(output_dir):
            os.makedirs(output_dir, exist_ok=True)
        
        # Save mesh to file
        file_ext = os.path.splitext(output_file)[1].lower()
        
        if file_ext == '.stl':
            mesh.save(output_file, binary=True)
        elif file_ext == '.obj':
            mesh.save(output_file)
        elif file_ext == '.ply':
            mesh.save(output_file)
        else:
            # Default to STL if extension is not recognized
            output_file = os.path.splitext(output_file)[0] + '.stl'
            mesh.save(output_file, binary=True)
        
        print(f"Mesh saved to {output_file}")
        return mesh
        
    except Exception as e:
        print(f"Error generating mesh: {str(e)}")
        return None

# Function to generate separate meshes for voids and particles
def generate_component_meshes(void_volume, particle_volume, output_dir):
    """
    Generate separate mesh files for voids and particles.
    
    Parameters:
    - void_volume: 3D array with voids marked as 1
    - particle_volume: 3D array with particles marked as 1
    - output_dir: Directory to save mesh files
    
    Returns:
    - Dictionary with mesh file paths
    """
    os.makedirs(output_dir, exist_ok=True)
    
    mesh_files = {}
    
    # Generate void mesh
    void_mesh_path = os.path.join(output_dir, "voids_mesh.stl")
    void_mesh = generate_mesh_from_volume(void_volume, void_mesh_path)
    if void_mesh:
        mesh_files['voids'] = void_mesh_path
    
    # Generate particle mesh
    particle_mesh_path = os.path.join(output_dir, "particles_mesh.stl")
    particle_mesh = generate_mesh_from_volume(particle_volume, particle_mesh_path)
    if particle_mesh:
        mesh_files['particles'] = particle_mesh_path
          # Generate combined sample mesh (matrix material)
    matrix_volume = ~(void_volume | particle_volume)
    matrix_mesh_path = os.path.join(output_dir, "matrix_mesh.stl")
    matrix_mesh = generate_mesh_from_volume(matrix_volume, matrix_mesh_path)
    if matrix_mesh:
        mesh_files['matrix'] = matrix_mesh_path
    
    # Generate complete sample mesh (all components together)
    complete_mesh_path = os.path.join(output_dir, "complete_sample.vtm")
    
    # Create multiple meshes in a single VTK file
    blocks = pv.MultiBlock()
    
    if void_mesh:
        blocks.append(void_mesh)
    if particle_mesh:
        blocks.append(particle_mesh)
    if matrix_mesh:
        blocks.append(matrix_mesh)
    
    # Only save if we have at least one valid mesh
    if len(blocks) > 0:
        blocks.save(complete_mesh_path)
        mesh_files['complete'] = complete_mesh_path
    
    return mesh_files

# Function to process a stack of images and create 3D visualizations and meshes
def process_image_stack(images, thresholds_dict, output_dir):
    """
    Process a stack of images to create 3D visualizations and meshes.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Base directory for outputs
    
    Returns:
    - Dictionary with results and file paths
    """
    try:
        # Create output subdirectories
        visualization_dir = os.path.join(output_dir, "3d_visualizations")
        mesh_dir = os.path.join(output_dir, "meshes")
        segmentation_dir = os.path.join(output_dir, "segmented_images")
        morphology_dir = os.path.join(output_dir, "void_morphology")
        
        for directory in [visualization_dir, mesh_dir, segmentation_dir, morphology_dir]:
            os.makedirs(directory, exist_ok=True)
        
        void_slices = []
        particle_slices = []
        segmented_images = []
        
        raster_void_count = 0
        spherical_void_count = 0
        large_void_count = 0
        
        # Lists to store void metrics for each slice
        void_morphology_metrics_per_slice = []
        void_details_list = []
        
        # Process each image
        for idx, (filename, img) in enumerate(images):
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                t1, t2 = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment voids and particles
            void_mask = segment_voids_with_t1(gray_img, t1)
            particle_mask = segment_particles_with_t2(gray_img, t2)
            
            # Annotate voids
            annotated_img, raster_count, spherical_count, large_count = annotate_voids(gray_img, void_mask)
            
            # Extract NND metrics
            nnd_metrics = extract_nnd_metrics(void_mask)
            if nnd_metrics:
                nnd_metrics['filename'] = filename
                nnd_metrics['slice_index'] = idx
                void_morphology_metrics_per_slice.append(nnd_metrics)
            
            # Extract void details for each individual void
            labels = measure.label(void_mask > 0)
            regions = measure.regionprops(labels)
            
            for i, region in enumerate(regions):
                y, x = region.centroid
                width = region.bbox[3] - region.bbox[1]
                height = region.bbox[2] - region.bbox[0]
                area = region.area
                perimeter = region.perimeter if hasattr(region, 'perimeter') else 0
                
                # Calculate circularity and aspect ratio
                circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
                aspect_ratio = max(width, height) / max(min(width, height), 1)
                
                # Calculate orientation
                orientation_rad = np.arctan2(height, width)
                orientation_deg = orientation_rad * 180 / np.pi
                
                # Store void details
                void_details = {
                    'Slice': idx + 1,
                    'Image': filename,
                    'ID': i + 1,
                    'X': x,
                    'Y': y,
                    'Width': width,
                    'Height': height,
                    'Area': area,
                    'Circularity': circularity,
                    'Aspect Ratio': aspect_ratio,
                    'Orientation_deg': orientation_deg
                }
                void_details_list.append(void_details)
            
            # Update counts
            raster_void_count += raster_count
            spherical_void_count += spherical_count
            large_void_count += large_count
            
            # Save segmented image
            segmented_path = os.path.join(segmentation_dir, f"segmented_{filename}")
            cv2.imwrite(segmented_path, annotated_img)
            segmented_images.append(segmented_path)
            
            # Add segmentation masks to slices for 3D reconstruction
            void_slices.append(void_mask)
            particle_slices.append(particle_mask)
        
        # Save void details to CSV
        if void_details_list:
            void_details_df = pd.DataFrame(void_details_list)
            void_details_csv_path = os.path.join(output_dir, "Void_Details.csv")
            void_details_df.to_csv(void_details_csv_path, index=False)
            print(f"Void details saved to: {void_details_csv_path}")
        
        # Check if we have any valid slices
        if not void_slices or not particle_slices:
            raise ValueError("No valid slices found for 3D reconstruction")
        
        # Create 3D volumes
        void_volume = create_3d_volume(void_slices)
        particle_volume = create_3d_volume(particle_slices)
        
        # Create 3D visualization - Fixed to handle tuple return
        viz_path = os.path.join(visualization_dir, "3d_segmentation.png")
        viz_result = create_professional_3d_visualization(
            void_volume, particle_volume, viz_path
        )
        
        if viz_result[0] is None:  # Check if visualization failed
            raise RuntimeError("3D visualization failed")
        
        viz_path, html_path = viz_result
        
        # Generate mesh files
        mesh_files = generate_component_meshes(void_volume, particle_volume, mesh_dir)
        
        # Calculate volume fractions with safe division
        total_volume = void_volume.size
        if total_volume > 0:
            void_volume_fraction = (np.sum(void_volume) / total_volume) * 100
            particle_volume_fraction = (np.sum(particle_volume) / total_volume) * 100
            matrix_volume_fraction = 100 - void_volume_fraction - particle_volume_fraction
        else:
            void_volume_fraction = 0
            particle_volume_fraction = 0
            matrix_volume_fraction = 0
        
        # Aggregate morphology metrics
        aggregate_void_metrics = {}
        
        if void_morphology_metrics_per_slice:
            # Calculate mean values across all slices
            for key in void_morphology_metrics_per_slice[0].keys():
                if (key not in ['filename', 'slice_index', 'nnd_values'] and 
                    isinstance(void_morphology_metrics_per_slice[0][key], (int, float)) and 
                    void_morphology_metrics_per_slice[0][key] is not None):
                    values = [metrics[key] for metrics in void_morphology_metrics_per_slice 
                             if key in metrics and metrics[key] is not None]
                    if values:
                        aggregate_void_metrics[key] = np.mean(values)
        
        # Calculate global morphology statistics from void details
        if void_details_list:
            void_df = pd.DataFrame(void_details_list)
            morphology_stats = {
                'circularity_mean': void_df['Circularity'].mean(),
                'circularity_std': void_df['Circularity'].std(),
                'aspect_ratio_mean': void_df['Aspect Ratio'].mean(),
                'aspect_ratio_std': void_df['Aspect Ratio'].std(),
                'elongated_void_percentage': (void_df['Aspect Ratio'] > 2).mean() * 100,
                'circular_void_percentage': (void_df['Circularity'] > 0.8).mean() * 100,
                'void_area_mean': void_df['Area'].mean(),
                'void_area_skewness': void_df['Area'].skew()
            }
            
            # Calculate orientation distribution
            if 'Orientation_deg' in void_df.columns:
                morphology_stats['orientation_0_30_pct'] = ((void_df['Orientation_deg'] >= 0) & 
                                                          (void_df['Orientation_deg'] < 30)).mean() * 100
                morphology_stats['orientation_30_60_pct'] = ((void_df['Orientation_deg'] >= 30) & 
                                                            (void_df['Orientation_deg'] < 60)).mean() * 100
                morphology_stats['orientation_60_90_pct'] = ((void_df['Orientation_deg'] >= 60) & 
                                                            (void_df['Orientation_deg'] <= 90)).mean() * 100
                
                # Calculate orientation tensor
                rad_angles = void_df['Orientation_deg'] * np.pi / 180
                nx_values = np.cos(rad_angles)
                ny_values = np.sin(rad_angles)
                
                orientation_tensor = np.zeros((2, 2))
                for nx, ny in zip(nx_values, ny_values):
                    n = np.array([nx, ny])
                    orientation_tensor += np.outer(n, n)
                
                if len(nx_values) > 0:
                    orientation_tensor /= len(nx_values)
                    eigenvalues, _ = np.linalg.eigh(orientation_tensor)
                    if (eigenvalues[1] + eigenvalues[0]) > 0:
                        morphology_stats['orientation_alignment'] = (eigenvalues[1] - eigenvalues[0]) / (eigenvalues[1] + eigenvalues[0])
            
            # Update aggregate metrics with morphology stats
            aggregate_void_metrics.update(morphology_stats)
        
        # Compile results
        results = {
            "raster_void_count": raster_void_count,
            "spherical_void_count": spherical_void_count,
            "large_void_count": large_void_count,
            "total_void_count": raster_void_count + spherical_void_count + large_void_count,
            "void_volume_fraction": void_volume_fraction,
            "particle_volume_fraction": particle_volume_fraction,
            "matrix_volume_fraction": matrix_volume_fraction,
            "visualization_path": viz_path,
            "html_visualization_path": html_path,
            "mesh_files": mesh_files,
            "segmented_images": segmented_images,
            "void_volume_shape": void_volume.shape,
            "physical_dimensions_mm": (
                void_volume.shape[2] * MICRONS_PER_PIXEL / 1000,
                void_volume.shape[1] * MICRONS_PER_PIXEL / 1000,
                void_volume.shape[0] * SLICE_THICKNESS_MICRONS / 1000
            ),
            "void_morphology_metrics": aggregate_void_metrics
        }
        
        # Save results to CSV
        results_csv_path = os.path.join(output_dir, "3d_analysis_results.csv")
        with open(results_csv_path, 'w', newline='') as csvfile:
            writer = csv.writer(csvfile)
            writer.writerow(['Parameter', 'Value', 'Units'])
            writer.writerow(['Raster Void Count', raster_void_count, ''])
            writer.writerow(['Spherical Void Count', spherical_void_count, ''])
            writer.writerow(['Large Void Count', large_void_count, ''])
            writer.writerow(['Total Void Count', results["total_void_count"], ''])
            writer.writerow(['Void Volume Fraction', f"{void_volume_fraction:.4f}", '%'])
            writer.writerow(['Particle Volume Fraction', f"{particle_volume_fraction:.4f}", '%'])
            writer.writerow(['Matrix Volume Fraction', f"{matrix_volume_fraction:.4f}", '%'])
            writer.writerow(['X Dimension', f"{results['physical_dimensions_mm'][0]:.4f}", 'mm'])
            writer.writerow(['Y Dimension', f"{results['physical_dimensions_mm'][1]:.4f}", 'mm'])
            writer.writerow(['Z Dimension', f"{results['physical_dimensions_mm'][2]:.4f}", 'mm'])
            
            # Add void morphology metrics to CSV
            writer.writerow(['', '', ''])  # Empty row as separator
            writer.writerow(['Void Morphology and Distribution Metrics', '', ''])
            for key, value in aggregate_void_metrics.items():
                # Format the key for better readability
                formatted_key = ' '.join(word.capitalize() for word in key.split('_'))
                if isinstance(value, (int, float)):
                    writer.writerow([formatted_key, f"{value:.4f}", ''])
        
        # Save detailed morphology metrics to CSV
        if void_morphology_metrics_per_slice:
            morphology_df = pd.DataFrame(void_morphology_metrics_per_slice)
            # Remove the nnd_values column which can be very large
            if 'nnd_values' in morphology_df.columns:
                morphology_df = morphology_df.drop(columns=['nnd_values'])
            morphology_csv_path = os.path.join(morphology_dir, "void_morphology_metrics.csv")
            morphology_df.to_csv(morphology_csv_path, index=False)
            print(f"Void morphology metrics saved to: {morphology_csv_path}")
        
        # Create visualization of void orientation distribution
        if void_details_list:
            void_df = pd.DataFrame(void_details_list)
            if 'Orientation_deg' in void_df.columns:
                # Create orientation rose diagram
                fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})
                
                # Convert to radians and bin the data
                orientations_rad = void_df['Orientation_deg'] * np.pi / 180
                bins = np.linspace(0, np.pi, 19)  # 18 bins covering 0 to 180 degrees
                
                # Create histogram
                hist, bin_edges = np.histogram(orientations_rad, bins=bins)
                
                # Plot rose diagram
                width = np.pi / 18
                bars = ax.bar(bin_edges[:-1], hist, width=width, alpha=0.7)
                
                # Customize appearance
                ax.set_theta_zero_location('N')  # Set 0 degrees at the top
                ax.set_theta_direction(-1)  # Clockwise
                ax.set_xticks(np.linspace(0, np.pi, 7))
                ax.set_xticklabels(['0°', '30°', '60°', '90°', '120°', '150°', '180°'])
                ax.set_title('Void Orientation Distribution', fontsize=16)
                ax.grid(True)
                
                # Save the figure
                orientation_viz_path = os.path.join(morphology_dir, 'void_orientation_distribution.png')
                plt.savefig(orientation_viz_path, dpi=300, bbox_inches='tight')
                plt.close(fig)
                print(f"Void orientation distribution visualization saved to: {orientation_viz_path}")
        
        return results

    except Exception as e:
        print(f"Error in image stack processing: {e}")
        import traceback
        traceback.print_exc()
        return None

# Additional helper functions for specialized analysis
def generate_raster_void_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing raster voids
    - Dictionary of raster void details
    """
    raster_void_slices = []
    raster_void_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            t1, _ = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment voids using T1 threshold
        void_mask = segment_voids_with_t1(gray_img, t1)
        
        # Classify voids to get specifically raster voids
        raster_voids, _, _ = classify_voids(void_mask)
        
        # Create a binary mask for raster voids
        raster_void_mask = np.zeros_like(gray_img)
        cv2.drawContours(raster_void_mask, raster_voids, -1, 255, -1)
        
        # Store slice and details
        raster_void_slices.append(raster_void_mask)
        
        # Collect void details
        void_details = {
            'filename': filename,
            'raster_void_count': len(raster_voids),
            'raster_void_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in raster_voids]
        }
        raster_void_details.append(void_details)
    
    # Create 3D volume
    raster_void_volume = create_3d_volume(raster_void_slices)
    
    return raster_void_volume, raster_void_details

def generate_particle_3d_volume(images, thresholds_dict):
    """
    Generate a 3D volume specifically for particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    
    Returns:
    - 3D numpy array representing particles
    - Dictionary of particle details
    """
    particle_slices = []
    particle_details = []

    for filename, img in images:
        # Convert to grayscale if necessary
        if len(img.shape) == 3:
            gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        else:
            gray_img = img
        
        # Get thresholds for this image
        if filename in thresholds_dict:
            _, t2 = thresholds_dict[filename]
        else:
            print(f"No thresholds found for {filename}, skipping.")
            continue
        
        # Segment particles using T2 threshold
        particle_mask = segment_particles_with_t2(gray_img, t2)
        
        # Find particle contours
        num_labels, labels_particle = cv2.connectedComponents(particle_mask)
        
        # Create list of particle contours
        particle_contours = []
        for label in range(1, num_labels):
            component_mask = (labels_particle == label).astype(np.uint8) * 255
            contours, _ = cv2.findContours(component_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            particle_contours.extend(contours)
        
        # Create binary mask for particles
        particle_binary = np.zeros_like(gray_img)
        cv2.drawContours(particle_binary, particle_contours, -1, 255, -1)
        
        # Store slice and details
        particle_slices.append(particle_binary)
        
        # Collect particle details
        particle_details.append({
            'filename': filename,
            'particle_count': len(particle_contours),
            'particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in particle_contours]
        })
    
    # Create 3D volume
    particle_volume = create_3d_volume(particle_slices)
    
    return particle_volume, particle_details

def generate_raster_void_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of raster voids.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "raster_void_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Generate raster void volume
        raster_void_volume, raster_void_details = generate_raster_void_3d_volume(images, thresholds_dict)
        
        # Visualize 3D volume of raster voids
        viz_path = os.path.join(visualization_dir, "raster_voids_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            raster_void_volume, 
            np.zeros_like(raster_void_volume),  # Empty particle volume
            viz_path
        )
        
        return viz_path, raster_void_volume

    except Exception as e:
        print(f"Error generating raster void 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def generate_spherical_particle_3d_representation(images, thresholds_dict, output_dir):
    """
    Generate a specialized 3D representation of spherical particles.
    
    Parameters:
    - images: List of (filename, image) tuples
    - thresholds_dict: Dictionary mapping filenames to thresholds
    - output_dir: Directory to save visualization
    
    Returns:
    - Tuple of (visualization path, 3D volume)
    """
    try:
        # Create output directory
        visualization_dir = os.path.join(output_dir, "spherical_particle_3d_representations")
        os.makedirs(visualization_dir, exist_ok=True)
        
        # Collect spherical particle slices
        spherical_particle_slices = []
        spherical_particle_details = []

        for filename, img in images:
            # Convert to grayscale if necessary
            if len(img.shape) == 3:
                gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            else:
                gray_img = img
            
            # Get thresholds for this image
            if filename in thresholds_dict:
                _, t2 = thresholds_dict[filename]
            else:
                print(f"No thresholds found for {filename}, skipping.")
                continue
            
            # Segment particles using T2 threshold
            particle_mask = segment_particles_with_t2(gray_img, t2)
            
            # Classify voids to extract specifically spherical particles
            _, spherical_particles, _ = classify_voids(particle_mask)
            
            # Create a binary mask for spherical particles
            spherical_particle_mask = np.zeros_like(gray_img)
            cv2.drawContours(spherical_particle_mask, spherical_particles, -1, 255, -1)
            
            # Store slice and details
            spherical_particle_slices.append(spherical_particle_mask)
            
            # Collect particle details
            particle_details = {
                'filename': filename,
                'spherical_particle_count': len(spherical_particles),
                'spherical_particle_areas': [cv2.contourArea(contour) * (MICRONS_PER_PIXEL ** 2) for contour in spherical_particles]
            }
            spherical_particle_details.append(particle_details)
        
        # Create 3D volume
        spherical_particle_volume = create_3d_volume(spherical_particle_slices)
        
        # Visualize 3D volume of spherical particles
        viz_path = os.path.join(visualization_dir, "spherical_particles_3d_representation.png")
        viz_path, _ = create_professional_3d_visualization(
            np.zeros_like(spherical_particle_volume),  # Empty void volume
            spherical_particle_volume, 
            viz_path
        )
        
        return viz_path, spherical_particle_volume

    except Exception as e:
        print(f"Error generating spherical particle 3D representation: {e}")
        import traceback
        traceback.print_exc()
        return None, None

def visualize_3d_volumes(raster_void_volume, particle_volume, output_dir):
    """
    Visualize and save 3D volumes for raster voids and particles.
    
    Parameters:
    - raster_void_volume: 3D numpy array of raster voids
    - particle_volume: 3D numpy array of particles
    - output_dir: Directory to save visualizations
    
    Returns:
    - Dictionary with visualization paths and details
    """
    # Create output directories
    visualization_dir = os.path.join(output_dir, "3d_component_visualizations")
    mesh_dir = os.path.join(output_dir, "component_meshes")
    
    os.makedirs(visualization_dir, exist_ok=True)
    os.makedirs(mesh_dir, exist_ok=True)
    
    # Visualization for raster voids
    raster_void_viz_path = os.path.join(visualization_dir, "raster_voids_3d.png")
    raster_void_viz_path, _ = create_professional_3d_visualization(
        raster_void_volume, 
        np.zeros_like(raster_void_volume),  # Empty particle volume
        raster_void_viz_path
    )
    
    # Visualization for particles
    particle_viz_path = os.path.join(visualization_dir, "particles_3d.png")
    particle_viz_path, _ = create_professional_3d_visualization(
        np.zeros_like(particle_volume),  # Empty void volume
        particle_volume, 
        particle_viz_path
    )
    
    # Generate mesh files
    raster_void_meshes = generate_component_meshes(
        raster_void_volume, 
        np.zeros_like(raster_void_volume), 
        os.path.join(mesh_dir, "raster_void_meshes")
    )
    
    particle_meshes = generate_component_meshes(
        np.zeros_like(particle_volume),
        particle_volume, 
        os.path.join(mesh_dir, "particle_meshes")
    )
    
    # Calculate volume fractions
    total_raster_void_volume = raster_void_volume.size
    raster_void_volume_fraction = (np.sum(raster_void_volume) / total_raster_void_volume) * 100
    
    total_particle_volume_size = particle_volume.size
    particle_volume_fraction = (np.sum(particle_volume) / total_particle_volume_size) * 100
    
    # Compile results
    results = {
        "raster_void_visualization_path": raster_void_viz_path,
        "particle_visualization_path": particle_viz_path,
        "raster_void_meshes": raster_void_meshes,
        "particle_meshes": particle_meshes,
        "raster_void_volume_fraction": raster_void_volume_fraction,
        "particle_volume_fraction": particle_volume_fraction
    }
    
    return results

if __name__ == "__main__":
    root = tk.Tk()
    app = FoamAnalyzerGUI(root)
    root.mainloop()
```
